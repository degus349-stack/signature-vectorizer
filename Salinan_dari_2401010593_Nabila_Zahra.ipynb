```python
# Import library
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload gambar
uploaded = files.upload()
image_path = list(uploaded.keys())[0]

# Baca gambar
img = cv2.imread(image_path)

# Konversi ke RGB (agar tampil normal di matplotlib)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Fungsi Gamma Correction
def gamma_correction(image, gamma):
    # 1. Normalisasi (0-255 → 0-1)
    normalized = image / 255.0

    # 2. Transformasi (Power Law)
    corrected = np.power(normalized, gamma)

    # 3. Denormalisasi (0-1 → 0-255)
    result = np.uint8(corrected * 255)

    return result

# Coba beberapa nilai gamma
gamma1 = 0.5   # lebih terang
gamma2 = 1.5   # lebih gelap

img_gamma1 = gamma_correction(img_rgb, gamma1)
img_gamma2 = gamma_correction(img_rgb, gamma2)

# Tampilkan hasil
plt.figure(figsize=(12,6))

plt.subplot(1,3,1)
plt.imshow(img_rgb)
plt.title("Gambar Asli")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(img_gamma1)
plt.title("Gamma = 0.5 (Lebih Terang)")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(img_gamma2)
plt.title("Gamma = 1.5 (Lebih Gelap)")
plt.axis('off')

plt.show()
```

```python
# =========================
# UTS SETELAH REVISI
# =========================
import gradio as gr
import cv2
import numpy as np
import matplotlib.pyplot as plt

# =========================
# MATRIX VISUALIZATION
# =========================
def matrix_to_image_with_text(matrix):
    matrix = np.array(matrix)

    norm = cv2.normalize(matrix, None, 0, 255, cv2.NORM_MINMAX)
    norm = np.uint8(norm)

    scale = 50
    h, w = norm.shape
    img = cv2.resize(norm, (w*scale, h*scale), interpolation=cv2.INTER_NEAREST)

    # ubah ke BGR biar bisa warna
    img = cv2.cvtColor(img, cv2.COLOR_GRAY2BGR)

    for i in range(h):
        for j in range(w):
            val = str(matrix[i][j])
            cv2.putText(img, val,
                        (j*scale + 5, i*scale + 30),
                        cv2.FONT_HERSHEY_SIMPLEX,
                        0.5,
                        (255, 0, 0),  # 🔵 BIRU (BGR)
                        1)

    return img

# =========================
# MATRIX before after
# =========================
def get_matrix_pair(img_before, img_after):
    before = img_before[0:5,0:5]
    after = img_after[0:5,0:5]

    before_img = matrix_to_image_with_text(before[:,:,0])
    after_img = matrix_to_image_with_text(after[:,:,0])

    return before_img, after_img


# =========================
# HISTOGRAM RGB
# =========================
def get_histogram_rgb(image, title="Histogram", filename="hist.png"):
    color = ('b','g','r')

    plt.figure()
    for i,col in enumerate(color):
        hist = cv2.calcHist([image],[i],None,[256],[0,256])
        plt.plot(hist, color=col)

    plt.title(title)
    plt.xlabel("Intensitas")
    plt.ylabel("Jumlah Pixel")

    plt.savefig(filename)
    plt.close()

    return filename

# =========================
# DASAR (GRAYSCALE)
# =========================
def grayscale_process(image):
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    h, w = image.shape[:2]

    before_img, after_img = get_matrix_pair(image, cv2.cvtColor(gray, cv2.COLOR_GRAY2BGR))

    r,g,b = image[0,0]
    hasil = int(0.299*r + 0.587*g + 0.114*b)

    return (
        gray,
        f"{w} x {h}",
        before_img,
        after_img,
        "Gray = 0.299R + 0.587G + 0.114B (standar luminance Rec.601)",
        f"({r},{g},{b}) → {hasil}",
        "Konversi RGB ke grayscale"
    )

# =========================
# ARITMATIKA
# =========================
def proses_arit(img1, img2, op):
    img2 = cv2.resize(img2,(img1.shape[1],img1.shape[0]))
    a,b = img1[0,0][0], img2[0,0][0]

    if op == "Penjumlahan":
        res = cv2.add(img1,img2)
        teori = "S = A + B"
        rumus_img = f"{a} + {b}"
        langkah = f"{a} + {b} = {min(a+b,255)}"

    elif op == "Pengurangan":
        res = cv2.subtract(img1,img2)
        teori = "S = A - B"
        rumus_img = f"{a} - {b}"
        langkah = f"{a} - {b} = {max(a-b,0)}"

    elif op == "Perkalian":
        res = cv2.multiply(img1,img2)
        teori = "S = A × B"
        rumus_img = f"{a} × {b}"
        langkah = f"{a} × {b} = {a*b}"

    elif op == "Pembagian":
        img2 = np.where(img2==0,1,img2)
        res = cv2.divide(img1,img2)
        teori = "S = A / B"
        rumus_img = f"{a} / {b}"
        langkah = f"{round(a/b,2)}"

    hist_before = get_histogram_rgb(img1, "Histogram Before", "before.png")
    hist_after = get_histogram_rgb(res, "Histogram After", "after.png")

    before_img, after_img = get_matrix_pair(img1, res)

    return res, hist_before, hist_after, before_img, after_img, teori, rumus_img, langkah


# =========================
# GEOMETRI
# =========================
def proses_geo(img, op, tx, ty, angle, scale):
    before = img.copy()
    h,w = img.shape[:2]

    if op == "Translasi":
        M = np.float32([[1,0,tx],[0,1,ty]])
        res = cv2.warpAffine(img,M,(w,h))
        teori = "x' = x + tx, y' = y + ty"
        rumus_img = f"(0,0) → ({tx},{ty})"
        langkah = "Perpindahan posisi piksel"

    elif op == "Rotasi":
        M = cv2.getRotationMatrix2D((w//2,h//2),angle,1)
        res = cv2.warpAffine(img,M,(w,h))
        teori = "Rotasi terhadap pusat"
        rumus_img = f"Sudut = {angle}°"
        langkah = "Rotasi citra"

    elif op == "Scaling":
        scaled = cv2.resize(img, None, fx=scale, fy=scale)

        # simpan ukuran asli hasil scaling
        real_h, real_w = scaled.shape[:2]

        canvas = np.zeros_like(img)

        sh, sw = scaled.shape[:2]
        y_offset = max((h - sh) // 2, 0)
        x_offset = max((w - sw) // 2, 0)

        if scale < 1:
            canvas[y_offset:y_offset+sh, x_offset:x_offset+sw] = scaled
            res = canvas
        else:
            res = scaled[
                (sh - h)//2:(sh - h)//2 + h,
                (sw - w)//2:(sw - w)//2 + w
            ]

        teori = "x' = x × s, y' = y × s"
        rumus_img = f"Scale = {scale}"
        langkah = "Zoom in/out"

        # 🔥 override size info khusus scaling
        size_info = f"{w}x{h} → {real_w}x{real_h}"


    elif op == "Flip Horizontal":
        res = cv2.flip(img,1)
        teori = "Refleksi horizontal"
        rumus_img = "Flip kiri-kanan"
        langkah = "Dibalik horizontal"

    elif op == "Flip Vertical":
        res = cv2.flip(img,0)
        teori = "Refleksi vertikal"
        rumus_img = "Flip atas-bawah"
        langkah = "Dibalik vertikal"

    before_img, after_img = get_matrix_pair(before, res)
    if op != "Scaling":
        size_info = f"{before.shape[1]}x{before.shape[0]} → {res.shape[1]}x{res.shape[0]}"

    return res, before_img, after_img, size_info, teori, rumus_img, langkah

# =========================
# TITIK
# =========================
def proses_titik(img, op, val, c):
    a = img[0,0][0]

    if op == "Brightness":
        res = cv2.convertScaleAbs(img, beta=val)
        teori = "s = r + b"
        rumus_img = f"{a} + {val}"
        langkah = f"{min(a+val,255)}"

    elif op == "Contrast":
        res = cv2.convertScaleAbs(img, alpha=c)
        teori = "s = r × c"
        rumus_img = f"{a} × {c}"
        langkah = f"{int(a*c)}"

    elif op == "Negasi":
        gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
        neg = 255 - gray
        res = cv2.cvtColor(neg, cv2.COLOR_GRAY2BGR)
        teori = "s = 255 - r"
        rumus_img = f"255 - {a}"
        langkah = f"{255-a}"

    elif op == "Threshold":
        gray = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
        _, thresh = cv2.threshold(gray,127,255,cv2.THRESH_BINARY)
        res = cv2.cvtColor(thresh, cv2.COLOR_GRAY2BGR)
        teori = "Setiap pixel dibandingkan dengan nilai batas (threshold). Kalau nilainya lebih besar dari batas, maka dijadikan putih (255). Kalau lebih kecil atau sama, dijadikan hitam (0)"
        rumus_img = f"{a} > 127"
        langkah = f"{255 if a>127 else 0}"

    before_img, after_img = get_matrix_pair(img, res)

    hist_before = get_histogram_rgb(img, "Before", "t_before.png")
    hist_after = get_histogram_rgb(res, "After", "t_after.png")

    return res, before_img, after_img, hist_before, hist_after, teori, rumus_img, langkah

# =========================
# SPASIAL (SOBEL)
# =========================
def proses_spasial(img):
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    sx = cv2.Sobel(gray, cv2.CV_64F, 1, 0)
    sy = cv2.Sobel(gray, cv2.CV_64F, 0, 1)

    res = np.sqrt(sx**2 + sy**2)
    res = np.uint8(np.absolute(res))

    # =========================
    # MATRIX BEFORE (PIXEL)
    # =========================
    sample = gray[0:3,0:3]
    sample_img = matrix_to_image_with_text(sample)

    # =========================
    # KERNEL VISUAL
    # =========================
    kx = np.array([[-1,0,1],
                   [-2,0,2],
                   [-1,0,1]])

    ky = np.array([[-1,-2,-1],
                   [ 0, 0, 0],
                   [ 1, 2, 1]])

    kx_img = matrix_to_image_with_text(kx)
    ky_img = matrix_to_image_with_text(ky)

    # =========================
    # AFTER (HASIL EDGE)
    # =========================
    after = res[0:3,0:3]
    after_img = matrix_to_image_with_text(after)

    teori = "G = √(Gx² + Gy²)"
    rumus_img = "Menggabungkan Sobel X dan Y"
    langkah = "Konvolusi pixel dengan kernel → menghasilkan edge"

    return res, sample_img, after_img, kx_img, ky_img, teori, rumus_img, langkah

# =========================
# STUDI KASUS (GABUNGAN)
# =========================
def studi_kasus(img, brightness_val, contrast_val, flip_type):
    before = img.copy()
    proses = img.copy()

    # 1. Brightness
    proses = cv2.convertScaleAbs(proses, beta=brightness_val)

    # 2. Contrast
    proses = cv2.convertScaleAbs(proses, alpha=contrast_val)

    # 3. Flip
    if flip_type == "Horizontal":
        out1 = cv2.flip(proses, 1)
        out2 = None

    elif flip_type == "Vertical":
        out1 = cv2.flip(proses, 0)
        out2 = None

    elif flip_type == "Keduanya":
        out1 = cv2.flip(proses, 1)
        out2 = cv2.flip(proses, 0)

    # ✅ MATRIX (ambil dari hasil pertama)
    before_img, after_img = get_matrix_pair(before, out1)

    teori = """Gabungan Operasi:
1. Brightness → s = r + b
2. Contrast → s = r × c
3. Flip → perubahan posisi pixel"""

    rumus = f"""
Brightness: +{brightness_val}
Contrast: ×{contrast_val}
Flip: {flip_type}
"""

    langkah = """Langkah:
1. Gambar diperbesar kecerahannya
2. Kontras diperkuat
3. Gambar dibalik sesuai pilihan
4. Semua proses menghasilkan 1/2 citra baru"""

    return out1, out2, before_img, after_img, teori, rumus, langkah

# =========================
# UI
# =========================
with gr.Blocks() as demo:
    gr.Markdown("# Aplikasi Citra Kelompok D")

    # DASAR (TETAP)
    with gr.Tab("Dasar"):
        img = gr.Image(label="Upload Gambar")
        btn = gr.Button("Proses Grayscale")

        out = gr.Image(label="Hasil")
        res = gr.Textbox(label="Resolusi")

        before = gr.Image(label="Matrix Before")
        after = gr.Image(label="Matrix After")
        teori = gr.Textbox(label="Rumus Umum")
        rumus = gr.Textbox(label="Rumus Berdasarkan Gambar")
        langkah = gr.Textbox(label="Langkah Perhitungan")

        btn.click(grayscale_process,img,[out,res,before,after,teori,rumus,langkah])

    # ARITMATIKA (HISTOGRAM)
    with gr.Tab("Aritmatika"):
        i1 = gr.Image(label="Gambar 1")
        i2 = gr.Image(label="Gambar 2")
        op = gr.Dropdown(["Penjumlahan","Pengurangan","Perkalian","Pembagian"], label="Pilih Operasi")
        btn = gr.Button("Proses")

        out = gr.Image(label="Hasil")
        hist1 = gr.Image(label="Histogram Before")
        hist2 = gr.Image(label="Histogram After")
        before = gr.Image(label="Matrix Before")
        after = gr.Image(label="Matrix After")
        teori = gr.Textbox(label="Rumus Umum")
        rumus = gr.Textbox(label="Rumus Berdasarkan Gambar")
        langkah = gr.Textbox(label="Langkah Perhitungan")

        btn.click(proses_arit,[i1,i2,op],[out,hist1,hist2,before,after,teori,rumus,langkah])

    # GEOMETRI (BEFORE AFTER)
    with gr.Tab("Geometri"):
        img = gr.Image(label="Upload Gambar")
        op = gr.Dropdown(["Translasi","Rotasi","Scaling","Flip Horizontal","Flip Vertical"])

        tx = gr.Slider(-100,100,0,label="Geser X (Translasi)")
        ty = gr.Slider(-100,100,0,label="Geser Y (Translasi)")
        angle = gr.Slider(0,180,45,label="Sudut Rotasi")
        scale = gr.Slider(0.5,2,1,label="Skala (Scaling)")

        def toggle_geo(op):
            return (
                gr.update(visible=(op=="Translasi")),
                gr.update(visible=(op=="Translasi")),
                gr.update(visible=(op=="Rotasi")),
                gr.update(visible=(op=="Scaling"))
            )

        op.change(toggle_geo, op, [tx,ty,angle,scale])
        demo.load(toggle_geo, op, [tx,ty,angle,scale])

        btn = gr.Button("Proses")

        out = gr.Image(label="After")
        teori = gr.Textbox(label="Matriks")
        before = gr.Image(label="Matrix Before")
        after = gr.Image(label="Matrix After")
        size = gr.Textbox(label="Ukuran Gambar")

        rumus = gr.Textbox(label="Rumus Berdasarkan Gambar")
        langkah = gr.Textbox(label="Langkah Perhitungan")

        btn.click(proses_geo,
          [img,op,tx,ty,angle,scale],
          [out,before,after,size,teori,rumus,langkah])

    # TITIK (MATRIX BEFORE AFTER)
    with gr.Tab("Operasi Titik"):
        img = gr.Image(label="Upload Gambar")
        op = gr.Dropdown(["Brightness","Contrast","Negasi","Threshold"], label="Pilih Operasi")

        val = gr.Slider(-100,100,0,visible=False,label="Brightness")
        c = gr.Slider(0.5,3,1,visible=False,label="Contrast")

        def toggle_titik(op):
            return (
                gr.update(visible=(op=="Brightness")),
                gr.update(visible=(op=="Contrast"))
            )

        op.change(toggle_titik, op, [val,c])
        demo.load(toggle_titik, op, [val,c])

        btn = gr.Button("Proses")

        out = gr.Image(label="Hasil")
        before = gr.Image(label="Matrix Before")
        after = gr.Image(label="Matrix After")
        hist1 = gr.Image(label="Histogram Before")
        hist2 = gr.Image(label="Histogram After")
        teori = gr.Textbox(label="Rumus Umum")
        rumus = gr.Textbox(label="Rumus Berdasarkan Gambar")
        langkah = gr.Textbox(label="Langkah Perhitungan")

        btn.click(proses_titik,[img,op,val,c],[out,before,after,hist1, hist2,teori,rumus,langkah])

    # SPASIAL (TETAP)
    with gr.Tab("Spasial (Sobel)"):
        img = gr.Image(label="Upload Gambar")
        btn = gr.Button("Proses Sobel")

        out = gr.Image(label="Hasil Edge")
        mat_before = gr.Image(label="Matriks Pixel (Before)")
        mat_after = gr.Image(label="Matriks Hasil Edge")

        kx = gr.Image(label="Kernel Sobel X")
        ky = gr.Image(label="Kernel Sobel Y")
        teori = gr.Textbox(label="Rumus Umum")
        rumus = gr.Textbox(label="Penjelasan")
        langkah = gr.Textbox(label="Langkah Perhitungan")

        btn.click(proses_spasial,img,[out, mat_before, mat_after, kx, ky, teori, rumus, langkah])

    # STUDI KASUS
    with gr.Tab("Studi Kasus"):
        img = gr.Image(label="Upload Gambar")

        brightness_val = gr.Slider(-100, 100, 20, label="Brightness (Pencerahan)")
        contrast_val = gr.Slider(0.5, 3, 1.2, label="Contrast (Kontras)")
        flip_type = gr.Dropdown(
            ["Horizontal", "Vertical", "Keduanya"],
            label="Jenis Flip"
        )

        btn = gr.Button("Proses Gabungan")
        out1 = gr.Image(label="Hasil 1")
        out2 = gr.Image(label="Hasil 2", visible=False)
        before = gr.Image(label="Matrix Before")
        after = gr.Image(label="Matrix After")
        teori = gr.Textbox(label="Rumus Umum", lines=3)
        def toggle_flip(flip_type):
            return gr.update(visible=(flip_type == "Keduanya"))

        flip_type.change(toggle_flip, flip_type, out2)


        rumus = gr.Textbox(label="Pengaturan yang Digunakan", lines=3)
        langkah = gr.Textbox(label="Langkah Proses", lines=4)

        btn.click(studi_kasus,[img, brightness_val, contrast_val, flip_type],[out1, out2, before, after, teori, rumus, langkah])

demo.launch()
```

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
from google.colab import files

# Upload gambar
uploaded = files.upload()

# Ambil nama file otomatis
filename = list(uploaded.keys())[0]

# Baca gambar dalam grayscale
im_in = cv2.imread(filename, cv2.IMREAD_GRAYSCALE)

# Cek gambar
if im_in is None:
    print("Gambar tidak berhasil dibaca!")
else:
    # Threshold OTSU
    th, im_th = cv2.threshold(
        im_in, 0, 255,
        cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU
    )

    # Copy image
    im_floodfill = im_th.copy()

    # Mask untuk floodFill
    h, w = im_th.shape[:2]
    mask = np.zeros((h+2, w+2), np.uint8)

    # Flood fill dari titik (0,0)
    cv2.floodFill(im_floodfill, mask, (0,0), 255)

    # Invert hasil floodfill
    im_floodfill_inv = cv2.bitwise_not(im_floodfill)

    # Gabungkan untuk mendapatkan foreground
    im_out = im_th | im_floodfill_inv

    # Tampilkan hasil dengan matplotlib
    plt.figure(figsize=(12,8))

    plt.subplot(221)
    plt.imshow(im_th, cmap='gray')
    plt.title("Thresholded")
    plt.xticks([]), plt.yticks([])

    plt.subplot(222)
    plt.imshow(im_floodfill, cmap='gray')
    plt.title("Floodfilled")
    plt.xticks([]), plt.yticks([])

    plt.subplot(223)
    plt.imshow(im_floodfill_inv, cmap='gray')
    plt.title("Inverted Floodfill")
    plt.xticks([]), plt.yticks([])

    plt.subplot(224)
    plt.imshow(im_out, cmap='gray')
    plt.title("Foreground Result")
    plt.xticks([]), plt.yticks([])

    plt.show()
```

```python
#PENIPISAN/Thinning

import cv2
import numpy as np
from matplotlib import pyplot as plt
from google.colab import files

# Upload gambar
uploaded = files.upload()

# Ambil nama file otomatis
filename = list(uploaded.keys())[0]

# Baca gambar sebagai grayscale
img = cv2.imread(filename, 0)

if img is None:
    print("Gambar tidak berhasil dibaca!")
else:
    # Threshold
    ret, img = cv2.threshold(img, 127, 255, 0)

    # Skeleton kosong
    skel = np.zeros(img.shape[:2], np.uint8)

    # Kernel cross 3x3
    element = cv2.getStructuringElement(cv2.MORPH_CROSS, (3,3))

    # Proses skeleton
    while True:
        # Opening
        opened = cv2.morphologyEx(img, cv2.MORPH_OPEN, element)

        # Selisih
        temp = cv2.subtract(img, opened)

        # Tambahkan ke skeleton
        skel = cv2.bitwise_or(skel, temp)

        # Erosi
        img = cv2.erode(img, element)

        # Jika sudah habis, stop
        if cv2.countNonZero(img) == 0:
            break

    # Tampilkan hasil
    plt.figure(figsize=(10,5))

    plt.subplot(121)
    plt.imshow(cv2.imread(filename,0), cmap='gray')
    plt.title("Original")
    plt.xticks([]), plt.yticks([])

    plt.subplot(122)
    plt.imshow(skel, cmap='gray')
    plt.title("Thinning")
    plt.xticks([]), plt.yticks([])

    plt.show()
```

```python
!pip install opencv-python-headless scikit-image matplotlib svgwrite -q

from google.colab import files
import cv2
import numpy as np
import matplotlib.pyplot as plt
import svgwrite
from IPython.display import SVG, display
from skimage import morphology
import zipfile

# =========================
# 1. Upload gambar
# =========================
uploaded = files.upload()
file_name = list(uploaded.keys())[0]
print("Memproses:", file_name)

img = cv2.imread(file_name)
if img is None:
    raise Exception("Gagal membaca gambar.")

original = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# =========================
# 2. Grayscale + Contrast
# =========================
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

p2, p98 = np.percentile(gray, 2), np.percentile(gray, 98)

if p98 > p2:
    gray = np.clip(
        (gray - p2) * 255.0 / (p98 - p2),
        0,
        255
    ).astype(np.uint8)

blur = cv2.GaussianBlur(gray, (5, 5), 0)

# =========================
# 3. Threshold Otsu
# =========================
_, binary = cv2.threshold(
    blur,
    0,
    255,
    cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU
)

# =========================
# 4. Cleaning dengan skimage
# =========================
mask = binary > 0



# Hapus titik/noise kecil
mask = morphology.remove_small_objects(mask, min_size=80)




binary_clean = (mask * 255).astype(np.uint8)

# =========================
# 5. Crop tanda tangan
# =========================
coords = cv2.findNonZero(binary_clean)

if coords is None:
    raise Exception("Tanda tangan tidak terdeteksi.")

x, y, w, h = cv2.boundingRect(coords)

padding = 20
x = max(0, x - padding)
y = max(0, y - padding)
w = min(w + padding * 2, binary_clean.shape[1] - x)
h = min(h + padding * 2, binary_clean.shape[0] - y)

cropped = binary_clean[y:y+h, x:x+w]

# =========================
# 6. Resize ke 400x400
# =========================
target = 400

scale = min(target / cropped.shape[1], target / cropped.shape[0])
new_w = int(cropped.shape[1] * scale)
new_h = int(cropped.shape[0] * scale)

resized = cv2.resize(
    cropped,
    (new_w, new_h),
    interpolation=cv2.INTER_AREA
)

_, resized = cv2.threshold(
    resized,
    127,
    255,
    cv2.THRESH_BINARY
)

canvas = np.zeros((target, target), dtype=np.uint8)

x_offset = (target - new_w) // 2
y_offset = (target - new_h) // 2

canvas[y_offset:y_offset+new_h, x_offset:x_offset+new_w] = resized

# PNG: background putih, tanda tangan hitam
final_png = cv2.bitwise_not(canvas)
cv2.imwrite("signature_preprocessed.png", final_png)

# =========================
# 7. Vector ke SVG
# =========================
contours, _ = cv2.findContours(
    canvas,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_TC89_KCOS
)

dwg = svgwrite.Drawing(
    "signature_vectorized.svg",
    size=("400px", "400px"),
    profile="full"
)

dwg.add(
    dwg.rect(
        insert=(0, 0),
        size=("400px", "400px"),
        fill="white"
    )
)

for cnt in contours:
    area = cv2.contourArea(cnt)

    if area > 80:
        approx = cv2.approxPolyDP(
            cnt,
            epsilon=0.8,
            closed=True
        )

        points = [
            (int(p[0][0]), int(p[0][1]))
            for p in approx
        ]

        if len(points) >= 3:
            dwg.add(
                dwg.polygon(
                    points=points,
                    fill="black",
                    stroke="black",
                    stroke_width=0.3
                )
            )

dwg.save()

# =========================
# 8. Tampilkan hasil
# =========================
plt.figure(figsize=(16, 8))

plt.subplot(2, 3, 1)
plt.imshow(original)
plt.title("Original")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(binary, cmap="gray")
plt.title("Threshold")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(binary_clean, cmap="gray")
plt.title("Noise Removed + Hole Closed")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(canvas, cmap="gray")
plt.title("Canvas 400x400")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(final_png, cmap="gray")
plt.title("Final PNG")
plt.axis("off")

plt.tight_layout()
plt.show()

print("Preview SVG:")
display(SVG(filename="signature_vectorized.svg"))

# =========================
# 9. Download ZIP
# =========================
with zipfile.ZipFile("hasil_signature.zip", "w") as zipf:
    zipf.write("signature_preprocessed.png")
    zipf.write("signature_vectorized.svg")

files.download("hasil_signature.zip")
```

```python
# ==========================================================
# INSTALL
# ==========================================================

!pip install opencv-python-headless scikit-image matplotlib svgwrite gradio pandas -q

# ==========================================================
# IMPORT
# ==========================================================

import cv2
import gradio as gr
import numpy as np
import svgwrite
import zipfile
import pandas as pd
import io

from skimage import morphology

# ==========================================================
# MAIN FUNCTION
# ==========================================================

def process_signature(image):

    if image is None:
        return (None,)*9 # 9 Output

    # RGB -> BGR
    img = cv2.cvtColor(
        image,
        cv2.COLOR_RGB2BGR
    )

    # ======================================================
    # GRAYSCALE
    # ======================================================
    gray = cv2.cvtColor(
        img,
        cv2.COLOR_BGR2GRAY
    )

    # ======================================================
    # OTSU THRESHOLD
    # ======================================================
    _, thresh = cv2.threshold(
        gray,
        0,
        255,
        cv2.THRESH_BINARY + cv2.THRESH_OTSU
    )

    # ======================================================
    # SIGNATURE MASK
    # ======================================================
    mask = thresh == 0

    mask = morphology.remove_small_objects(
        mask,
        min_size=80
    )

    mask = morphology.binary_closing(
        mask,
        morphology.disk(2)
    )

    # ======================================================
    # CLEAN IMAGE
    # ======================================================
    clean = np.ones_like(thresh) * 255
    clean[mask] = 0

    cv2.imwrite(
        "signature_clean.png",
        clean
    )

    # ======================================================
    # TRANSPARENT PNG
    # ======================================================
    rgba = cv2.cvtColor(
        clean,
        cv2.COLOR_GRAY2RGBA
    )

    rgba[clean == 255] = [255, 255, 255, 0]
    rgba[clean == 0]   = [0, 0, 0, 255]

    cv2.imwrite(
        "signature_transparent.png",
        rgba
    )

    # ======================================================
    # AUTO CROP
    # ======================================================
    binary_crop = np.zeros_like(clean)
    binary_crop[clean == 0] = 255

    coords = cv2.findNonZero(binary_crop)

    if coords is None:
        return (None,)*9

    x,y,w,h = cv2.boundingRect(coords)

    padding = 20
    x_pad = max(0, x-padding)
    y_pad = max(0, y-padding)
    w_pad = min(w+padding*2, clean.shape[1]-x_pad)
    h_pad = min(h+padding*2, clean.shape[0]-y_pad)

    cropped = binary_crop[y_pad:y_pad+h_pad, x_pad:x_pad+w_pad]

    # ======================================================
    # NORMALIZE 400x400
    # ======================================================
    target = 400
    scale = min(target/cropped.shape[1], target/cropped.shape[0])

    new_w = int(cropped.shape[1]*scale)
    new_h = int(cropped.shape[0]*scale)

    resized = cv2.resize(
        cropped,
        (new_w,new_h),
        interpolation=cv2.INTER_AREA
    )

    _, resized = cv2.threshold(resized, 127, 255, cv2.THRESH_BINARY)

    canvas = np.zeros((target,target), dtype=np.uint8)

    x_offset = (target-new_w) // 2
    y_offset = (target-new_h) // 2

    canvas[y_offset:y_offset+new_h, x_offset:x_offset+new_w] = resized
    final_png = cv2.bitwise_not(canvas)

    cv2.imwrite("signature_preprocessed.png", final_png)

    # ======================================================
    # MATRIX PREVIEW (FULL MATRIX) UNTUK SETIAP TAHAP
    # ======================================================
    # Menggunakan np.savetxt dengan io.StringIO agar proses rendering matriks
    # berukuran besar sangat cepat dan efisien tanpa for-loop manual.
    def format_full_matrix(matrix_data):
        s = io.StringIO()
        np.savetxt(s, matrix_data, fmt='%3d')
        return s.getvalue()

    # Mengekstrak seluruh piksel pada masing-masing state gambar
    preview_gray = format_full_matrix(gray)
    preview_clean = format_full_matrix(clean)
    preview_rgba = format_full_matrix(rgba[:, :, 3]) # Alpha channel
    preview_final_matrix = format_full_matrix(canvas)

    # ======================================================
    # MATRIX CSV & TEXT COORDINATES EXPORT
    # ======================================================
    pd.DataFrame(canvas).to_csv("signature_matrix.csv", index=False)

    with open("matrix_coordinates.txt", "w") as f:
        height,width = canvas.shape
        for cy in range(height):
            for cx in range(width):
                f.write(f"({cx},{cy}) = {int(canvas[cy,cx])}\n")

    with open("signature_pixels.txt", "w") as f:
        for cy in range(height):
            for cx in range(width):
                if canvas[cy,cx] > 0:
                    f.write(f"({cx},{cy}) = {int(canvas[cy,cx])}\n")

    # ======================================================
    # SVG
    # ======================================================
    contours,_ = cv2.findContours(
        canvas,
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_TC89_KCOS
    )

    dwg = svgwrite.Drawing("signature_vectorized.svg", size=("400px","400px"), profile="full")

    for cnt in contours:
        area = cv2.contourArea(cnt)
        if area > 80:
            approx = cv2.approxPolyDP(cnt, epsilon=0.8, closed=True)
            points = [(int(p[0][0]), int(p[0][1])) for p in approx]
            if len(points) >= 3:
                dwg.add(dwg.polygon(points=points, fill="black", stroke="black", stroke_width=0.3))

    dwg.save()

    # ======================================================
    # ZIP EXPORT
    # ======================================================
    with zipfile.ZipFile("hasil_signature.zip", "w") as zipf:
        files = [
            "signature_clean.png", "signature_transparent.png",
            "signature_preprocessed.png", "signature_vectorized.svg",
            "signature_matrix.csv", "matrix_coordinates.txt", "signature_pixels.txt"
        ]
        for file in files:
            zipf.write(file)

    return (
        output_original := image,
        output_clean := "signature_clean.png",
        output_transparent := "signature_transparent.png",
        output_final := "signature_preprocessed.png",
        text_gray := preview_gray,
        text_clean := preview_clean,
        text_rgba := preview_rgba,
        text_final := preview_final_matrix,
        output_zip := "hasil_signature.zip"
    )

# ==========================================================
# UI DENGAN TATA LETAK RAPI (Gradio Blocks)
# ==========================================================

with gr.Blocks(theme=gr.themes.Soft(), title="Signature Data Extractor") as demo:

    gr.Markdown("# 🖋️ Signature Cleaner & Full Matrix Extractor")
    gr.Markdown("Unggah foto tanda tangan. Sistem akan mengekstrak dan menampilkan **seluruh nilai matriks piksel** dari gambar pada setiap tahapan.")

    with gr.Row():

        # --- KOLOM KIRI (Input & Kontrol) ---
        with gr.Column(scale=1):
            input_image = gr.Image(type="numpy", label="Upload Tanda Tangan")
            btn_process = gr.Button("🚀 Proses & Ekstrak Matriks", variant="primary")

            gr.Markdown("### 📥 Unduh Hasil Lengkap")
            output_zip = gr.File(label="Download Paket (.zip)")

            gr.Markdown("""
            *File ZIP berisi:*
            - Gambar: `clean.png`, `transparent.png`, `preprocessed.png`
            - Vektor: `vectorized.svg`
            - Data Array: `matrix.csv`, `coordinates.txt`, `pixels.txt`
            """)

        # --- KOLOM KANAN (Visualisasi Output) ---
        with gr.Column(scale=2):

            with gr.Tabs():

                # TAB 1: Hasil Gambar
                with gr.TabItem("🖼️ Visualisasi Gambar"):
                    with gr.Row():
                        output_original = gr.Image(label="1. Original (Input)")
                        output_clean = gr.Image(type="filepath", label="2. Clean (Background Putih)")
                    with gr.Row():
                        output_transparent = gr.Image(type="filepath", label="3. Transparent PNG")
                        output_final = gr.Image(type="filepath", label="4. Preprocessed (400x400)")

                # TAB 2: Perbandingan Matriks 2D
                with gr.TabItem("🔢 Data Matriks Utuh (Seluruh Piksel)"):
                    gr.Markdown("⚠️ *Perhatian: Karena menampilkan keseluruhan piksel gambar, kotak teks ini mungkin akan mengalami sedikit lag saat di-scroll jika resolusi gambar yang diunggah sangat besar.*")
                    with gr.Row():
                        text_gray = gr.Textbox(label="1. Grayscale (Intensitas Cahaya: 0-255)", lines=22)
                        text_clean = gr.Textbox(label="2. Clean Biner (Hitam=0, Putih=255)", lines=22)
                    with gr.Row():
                        text_rgba = gr.Textbox(label="3. Alpha Channel (Transparansi: Tembus=0, Solid=255)", lines=22)
                        text_final = gr.Textbox(label="4. Final Normalization (Biner Standar 400x400)", lines=22)

    # ------------------------------------------------------
    # EVENT LISTENER
    # ------------------------------------------------------
    btn_process.click(
        fn=process_signature,
        inputs=[input_image],
        outputs=[
            output_original,
            output_clean,
            output_transparent,
            output_final,
            text_gray,
            text_clean,
            text_rgba,
            text_final,
            output_zip
        ]
    )

demo.launch(share=True)
```

```python
import cv2
import gradio as gr
import numpy as np
import svgwrite
import zipfile
import pandas as pd
import io

from skimage import morphology

# ==========================================================
# MAIN FUNCTION
# ==========================================================

def process_signature(image):
    if image is None:
        return (None,) * 13

    img = cv2.cvtColor(image, cv2.COLOR_RGB2BGR)
    h_orig, w_orig = img.shape[:2]
    dim_orig = f"{w_orig} x {h_orig} px"

    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    _, thresh = cv2.threshold(
        gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU
    )

    mask = thresh == 0
    mask = morphology.remove_small_objects(mask, min_size=80)
    mask = morphology.binary_closing(mask, morphology.disk(2))

    clean = np.ones_like(thresh) * 255
    clean[mask] = 0
    h_clean, w_clean = clean.shape
    dim_clean = f"{w_clean} x {h_clean} px"

    cv2.imwrite("signature_clean.png", clean)

    rgba = cv2.cvtColor(clean, cv2.COLOR_GRAY2RGBA)
    rgba[clean == 255] = [255, 255, 255, 0]
    rgba[clean == 0]   = [0, 0, 0, 255]

    cv2.imwrite("signature_transparent.png", rgba)

    binary_crop = np.zeros_like(clean)
    binary_crop[clean == 0] = 255

    coords = cv2.findNonZero(binary_crop)
    if coords is None:
        return (None,) * 13

    x, y, w, h = cv2.boundingRect(coords)

    padding = 20
    x_pad = max(0, x - padding)
    y_pad = max(0, y - padding)
    w_pad = min(w + padding * 2, clean.shape[1] - x_pad)
    h_pad = min(h + padding * 2, clean.shape[0] - y_pad)

    cropped = binary_crop[y_pad:y_pad + h_pad, x_pad:x_pad + w_pad]
    h_crop, w_crop = cropped.shape
    dim_crop = f"{w_crop} x {h_crop} px"

    target = 400
    scale = min(target / cropped.shape[1], target / cropped.shape[0])

    new_w = int(cropped.shape[1] * scale)
    new_h = int(cropped.shape[0] * scale)

    resized = cv2.resize(
        cropped, (new_w, new_h), interpolation=cv2.INTER_AREA
    )
    _, resized = cv2.threshold(resized, 127, 255, cv2.THRESH_BINARY)

    canvas = np.zeros((target, target), dtype=np.uint8)
    x_offset = (target - new_w) // 2
    y_offset = (target - new_h) // 2

    canvas[y_offset:y_offset + new_h, x_offset:x_offset + new_w] = resized
    final_png = cv2.bitwise_not(canvas)
    dim_final = f"{target} x {target} px"

    cv2.imwrite("signature_preprocessed.png", final_png)

    def format_matrix_safe(matrix_data, max_dim=60):
        h_mat, w_mat = matrix_data.shape[:2]
        if h_mat > max_dim or w_mat > max_dim:
            start_h = h_mat // 2 - max_dim // 2
            start_w = w_mat // 2 - max_dim // 2
            sampled = matrix_data[start_h:start_h + max_dim, start_w:start_w + max_dim]
            header = f"# PREVIEW PUSAT MATRIKS ({max_dim}x{max_dim})\n"
        else:
            sampled = matrix_data
            header = ""
        s = io.StringIO()
        np.savetxt(s, sampled, fmt='%3d')
        return header + s.getvalue()

    preview_gray = format_matrix_safe(gray)
    preview_clean = format_matrix_safe(clean)
    preview_rgba = format_matrix_safe(rgba[:, :, 3])
    preview_final_matrix = format_matrix_safe(canvas)

    pd.DataFrame(canvas).to_csv("signature_matrix.csv", index=False)

    with open("matrix_coordinates.txt", "w") as f:
        height, width = canvas.shape
        for cy in range(height):
            for cx in range(width):
                f.write(f"({cx},{cy}) = {int(canvas[cy,cx])}\n")

    with open("signature_pixels.txt", "w") as f:
        for cy in range(height):
            for cx in range(width):
                if canvas[cy, cx] > 0:
                    f.write(f"({cx},{cy}) = {int(canvas[cy,cx])}\n")

    contours, _ = cv2.findContours(
        canvas, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_TC89_KCOS
    )
    dwg = svgwrite.Drawing("signature_vectorized.svg", size=("400px", "400px"), profile="full")
    for cnt in contours:
        area = cv2.contourArea(cnt)
        if area > 80:
            approx = cv2.approxPolyDP(cnt, epsilon=0.8, closed=True)
            points = [(int(p[0][0]), int(p[0][1])) for p in approx]
            if len(points) >= 3:
                dwg.add(dwg.polygon(points=points, fill="black", stroke="black", stroke_width=0.3))
    dwg.save()

    with zipfile.ZipFile("hasil_signature.zip", "w") as zipf:
        files = [
            "signature_clean.png", "signature_transparent.png",
            "signature_preprocessed.png", "signature_vectorized.svg",
            "signature_matrix.csv", "matrix_coordinates.txt", "signature_pixels.txt"
        ]
        for file in files:
            zipf.write(file)

    return (
        image, "signature_clean.png", "signature_transparent.png", "signature_preprocessed.png",
        dim_orig, dim_clean, dim_crop, dim_final,
        preview_gray, preview_clean, preview_rgba, preview_final_matrix,
        "hasil_signature.zip"
    )

# ==========================================================
# SPOTIFY THEME CONFIGURATION (CSS INJECTION)
# ==========================================================

spotify_css = """
/* 1. Global Dark Background Immersive & Font Fallback */
body, .gradio-container {
    background-color: #121212 !important;
    font-family: 'SpotifyMixUI', 'CircularSp', 'Helvetica Neue', Helvetica, Arial, sans-serif !important;
    color: #ffffff !important;
}

/* 2. Container Card & Group Elevation (Level 1 & 2) */
.gr-group, .gr-box, .gr-form, .gr-panel, div[class*="g-group"] {
    background-color: #181818 !important;
    border: none !important;
    border-radius: 8px !important;
    box-shadow: rgba(0,0,0,0.3) 0px 8px 8px !important;
    padding: 16px !important;
    margin-bottom: 12px;
}

/* Efek Hover Card seperti Playlist Spotify */
.gr-group:hover, div[class*="g-group"]:hover {
    background-color: #1f1f1f !important;
    transition: background-color 0.2s ease;
}

/* 3. Tombol Utama (Spotify Green Pill CTA) */
button.primary, button[class*="gr-button-primary"] {
    background-color: #1ed760 !important;
    color: #000000 !important;
    font-weight: 700 !important;
    text-transform: uppercase !important;
    letter-spacing: 1.8px !important;
    font-size: 14px !important;
    border-radius: 9999px !important;
    border: none !important;
    padding: 12px 24px !important;
    box-shadow: rgba(0,0,0,0.4) 0px 4px 12px !important;
    cursor: pointer !important;
}
button.primary:hover {
    transform: scale(1.04) !important;
    background-color: #1fdf64 !important;
    transition: all 0.2s ease;
}

/* 4. Input Teks, Berkas, dan Gambar Rekayasa Inset Border */
input[type="text"], textarea, .gr-input, div[class*="gr-box-unpadded"], .file-preview {
    background-color: #1f1f1f !important;
    color: #ffffff !important;
    border: none !important;
    border-radius: 4px !important;
    box-shadow: rgb(18,18,18) 0px 1px 0px, rgb(77,77,77) 0px 0px 0px 1px inset !important;
}
input[type="text"]:focus, textarea:focus {
    box-shadow: rgb(18,18,18) 0px 1px 0px, rgb(30,215,96) 0px 0px 0px 1px inset !important;
}

/* 5. Pengaturan Label & Teks Sekunder (Silver Muted) */
.gr-form label, span[class*="block-label"], p, .gr-markdown h3, .gr-markdown h4 {
    color: #b3b3b3 !important;
    font-weight: 600 !important;
}
.gr-markdown h1 {
    color: #ffffff !important;
    font-weight: 700 !important;
    font-size: 24px !important;
    border-bottom: 1px solid #1f1f1f;
    padding-bottom: 8px;
}

/* 6. Kotak Matriks Teks Monospace Khusus */
textarea {
    font-family: 'Courier New', Courier, monospace !important;
    font-size: 11px !important;
    line-height: 1.2 !important;
    background-color: #121212 !important;
    color: #1ed760 !important; /* Membuat kode matriks menyala hijau */
}
"""

# ==========================================================
# UI DENGAN SEKUENSAL PIPELINE (SPOTIFY DESIGN SYSTEM)
# ==========================================================

with gr.Blocks(css=spotify_css, title="Signature Feature Extraction Pipeline") as demo:

    gr.Markdown("# 🖋️ Signature Feature Extraction Pipeline")
    gr.Markdown("Alur visualisasi data tanda tangan sekuensial dari citra mentah menuju matriks normalisasi array.")

    with gr.Row():

        # --------------------------------------------------
        # KOLOM KIRI: INPUT & CONTROL PANEL
        # --------------------------------------------------
        with gr.Column(scale=1):
            gr.Markdown("### 📥 Input & Ekstraksi")
            input_image = gr.Image(type="numpy", label="Upload Citra Tanda Tangan")
            btn_process = gr.Button("🚀 Jalankan Komputasi Pipeline", variant="primary")

            gr.Markdown("### 📦 Hasil Ekstraksi Data Utuh")
            output_zip = gr.File(label="Unduh Paket Lengkap (.zip)")
            gr.Markdown("""
            *Materi Berkas ZIP:*
            * `signature_matrix.csv` (Matriks 400x400)
            * `signature_pixels.txt` (Koordinat Aktif Objek)
            * `signature_vectorized.svg` (Skalabilitas Vektor)
            """)

        # --------------------------------------------------
        # KOLOM KANAN: PIPELINE (VISUAL vs DATA MATRIKS)
        # --------------------------------------------------
        with gr.Column(scale=2):
            gr.Markdown("### ⚙️ Tahapan Komputasi Citra & Matriks Terkait")

            # --- TAHAP 1: INPUT ORIGINAL VS GRAYSCALE MATRIX ---
            with gr.Group():
                gr.Markdown("#### **Tahap 1: Analisis Awal Citra Masukan**")
                with gr.Row():
                    with gr.Column():
                        output_original = gr.Image(label="1a. Citra Asli (RGB)")
                        dim_out_orig = gr.Textbox(label="📐 Resolusi Asli", interactive=False)
                    with gr.Column():
                        text_gray = gr.Textbox(label="1b. Matriks Grayscale (0-255)", lines=10, max_lines=10)

            # --- TAHAP 2: CLEANING BACKGROUND VS BINARY MATRIX ---
            with gr.Group():
                gr.Markdown("#### **Tahap 2: Segmentasi Thresholding & Filtrasi Morfologi**")
                with gr.Row():
                    with gr.Column():
                        output_clean = gr.Image(type="filepath", label="2a. Citra Bersih (Latar Putih)")
                        dim_out_clean = gr.Textbox(label="📐 Resolusi Segmentasi", interactive=False)
                    with gr.Column():
                        text_clean = gr.Textbox(label="2b. Matriks Biner Bersih (0 & 255)", lines=10, max_lines=10)

            # --- TAHAP 3: AUTO CROP BOUNDING BOX ---
            with gr.Group():
                gr.Markdown("#### **Tahap 3: Lokalisasi & Deteksi Batas Objek**")
                with gr.Row():
                    with gr.Column():
                        output_transparent = gr.Image(type="filepath", label="3a. Hasil Mask Transparan PNG")
                    with gr.Column():
                        dim_out_crop = gr.Textbox(label="✂️ Dimensi Area Objek Aktif (Bounding Box)", interactive=False)
                        text_rgba = gr.Textbox(label="3b. Matriks Transparansi (Alpha Channel)", lines=7, max_lines=7)

            # --- TAHAP 4: NORMALISASI AKHIR VS MATRIKS CANVAS ---
            with gr.Group():
                gr.Markdown("#### **Tahap 4: Normalisasi Matriks Standar**")
                with gr.Row():
                    with gr.Column():
                        output_final = gr.Image(type="filepath", label="4a. Citra Hasil Akhir (400x400)")
                        dim_out_final = gr.Textbox(label="📐 Resolusi Normalisasi", interactive=False)
                    with gr.Column():
                        text_final = gr.Textbox(label="4b. Matriks Standar Canvas (Tanda Tangan=255)", lines=10, max_lines=10)

    # ------------------------------------------------------
    # EVENT LISTENER MAPPING
    # ------------------------------------------------------
    btn_process.click(
        fn=process_signature,
        inputs=[input_image],
        outputs=[
            output_original, output_clean, output_transparent, output_final,
            dim_out_orig, dim_out_clean, dim_out_crop, dim_out_final,
            text_gray, text_clean, text_rgba, text_final,
            output_zip
        ]
    )

demo.launch(share=True)
```

```python
import cv2
import gradio as gr
import numpy as np
import svgwrite
import zipfile
import pandas as pd
import io
import os
import tempfile

# ==========================================================
# CSS DARK MODE ELEGAN
# ==========================================================
CUSTOM_CSS = """
:root {
    --bg-main:    #0f1117;
    --bg-card:    #1a1d27;
    --bg-input:   #12151f;
    --border:     #2a2d3a;
    --accent:     #7c6fcd;
    --accent-dim: #3d3670;
    --text-main:  #e8e8f0;
    --text-muted: #7a7a9a;
    --text-label: #a0a0c0;
    --radius:     12px;
}
body, .gradio-container {
    background-color: var(--bg-main) !important;
    color: var(--text-main) !important;
    font-family: 'Inter', 'Segoe UI', sans-serif !important;
}
.app-header {
    text-align: center;
    padding: 36px 20px 24px;
    border-bottom: 1px solid var(--border);
    margin-bottom: 24px;
}
.app-header h1 { font-size: 26px; font-weight: 700; color: var(--text-main); margin: 0 0 6px; }
.app-header p  { color: var(--text-muted); font-size: 13px; margin: 0; }

.card {
    background: var(--bg-card);
    border: 1px solid var(--border);
    border-radius: var(--radius);
    padding: 18px 20px;
    margin-bottom: 14px;
}
.step-badge {
    display: inline-flex;
    align-items: center;
    gap: 6px;
    background: var(--accent-dim);
    color: #c4b8ff;
    font-size: 11px;
    font-weight: 600;
    letter-spacing: 0.8px;
    text-transform: uppercase;
    padding: 3px 10px;
    border-radius: 20px;
    margin-bottom: 5px;
}
.step-title { font-size: 15px; font-weight: 600; color: var(--text-main); margin: 0 0 3px; }
.step-desc  { font-size: 12px; color: var(--text-muted); margin: 0; line-height: 1.5; }

.warn-box {
    background: #2a1f0e;
    border: 1px solid #7a4f1a;
    border-radius: 8px;
    padding: 12px 16px;
    font-size: 13px;
    color: #f5c97a;
    margin-bottom: 14px;
    line-height: 1.6;
}
.ok-box {
    background: #0e2a1a;
    border: 1px solid #1a6640;
    border-radius: 8px;
    padding: 12px 16px;
    font-size: 13px;
    color: #6ee7a8;
    margin-bottom: 14px;
    line-height: 1.6;
}

.gr-button-primary {
    background: var(--accent) !important;
    border: none !important;
    color: #fff !important;
    border-radius: 8px !important;
    font-weight: 600 !important;
    padding: 12px 0 !important;
    transition: opacity 0.2s !important;
}
.gr-button-primary:hover { opacity: 0.85 !important; }

.gr-textbox textarea, .gr-textbox input {
    background: var(--bg-input) !important;
    border: 1px solid var(--border) !important;
    color: var(--text-main) !important;
    border-radius: 8px !important;
    font-family: 'JetBrains Mono', 'Courier New', monospace !important;
    font-size: 12px !important;
}
label span { color: var(--text-label) !important; font-size: 12px !important; font-weight: 500 !important; }
.gr-image  { border-radius: 8px !important; border: 1px solid var(--border) !important; background: #000 !important; }
.gr-file   { background: var(--bg-input) !important; border: 1px solid var(--border) !important; border-radius: 8px !important; }
.flow-arrow { text-align: center; color: var(--accent); font-size: 20px; margin: 2px 0; }
"""

# ==========================================================
# ANALISIS KUALITAS GAMBAR
# ==========================================================
def analyze_image_quality(gray, h, w):
    warnings = []
    suggestions = []

    mean_bright = np.mean(gray)
    if mean_bright < 80:
        warnings.append("⚠️  Gambar terlalu gelap (rata-rata kecerahan: {:.0f}/255)".format(mean_bright))
        suggestions.append("→ Foto ulang di tempat yang lebih terang, atau dekat lampu.")
    elif mean_bright > 230:
        warnings.append("⚠️  Gambar terlalu terang/overexposed (rata-rata: {:.0f}/255)".format(mean_bright))
        suggestions.append("→ Kurangi cahaya atau jangan terlalu dekat sumber cahaya.")

    std_dev = np.std(gray)
    if std_dev < 20:
        warnings.append("⚠️  Kontras sangat rendah (std: {:.1f}) — kemungkinan tanda tangan pensil tipis atau foto buram.".format(std_dev))
        suggestions.append("→ Gunakan pensil/pulpen yang lebih tebal, atau perbesar kontras saat foto.")

    lap_var = cv2.Laplacian(gray, cv2.CV_64F).var()
    if lap_var < 50:
        warnings.append("⚠️  Gambar buram / tidak fokus (sharpness score: {:.1f})".format(lap_var))
        suggestions.append("→ Foto dengan kamera lebih stabil, pastikan tanda tangan dalam fokus.")

    _, tmp = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    dark_ratio = np.sum(tmp == 0) / (h * w)
    if dark_ratio > 0.5:
        warnings.append("⚠️  Lebih dari 50% piksel terdeteksi gelap ({:.0f}%) — kemungkinan ada bayangan besar atau background kotor.".format(dark_ratio*100))
        suggestions.append("→ Foto di kertas putih bersih, hindari bayangan tangan/objek lain.")

    if w < 200 or h < 200:
        warnings.append("⚠️  Resolusi terlalu kecil ({}×{} px) — hasil vektorisasi bisa kasar.".format(w, h))
        suggestions.append("→ Upload foto dengan resolusi minimal 400×400 px.")

    detail = "Kecerahan rata-rata : {:.0f}/255\nKontras (std dev)   : {:.1f}\nSharpness score     : {:.1f}\nArea gelap          : {:.1f}%\nResolusi            : {}×{} px".format(
        mean_bright, std_dev, lap_var, dark_ratio*100, w, h
    )

    is_ok = len(warnings) == 0
    return warnings, suggestions, is_ok, detail


def build_diagnosis_html(warnings, suggestions, is_ok, detail):
    if is_ok:
        return """<div class="ok-box">✅ <strong>Gambar terdeteksi baik!</strong> Tidak ada masalah kualitas yang ditemukan. Pipeline dijalankan dengan pengaturan standar.<br><br><small style='opacity:0.7'>{}</small></div>""".format(detail.replace('\n','<br>'))

    items = ""
    for w, s in zip(warnings, suggestions):
        items += "<li style='margin-bottom:6px'>{}<br><span style='color:#a0c4a0'>{}</span></li>".format(w, s)

    return """<div class="warn-box">
    <strong>⚠️ Terdeteksi masalah pada gambar — pipeline tetap dijalankan dengan penyesuaian otomatis:</strong>
    <ul style='margin:8px 0 0 16px;padding:0'>{}</ul>
    <details style='margin-top:10px;cursor:pointer'>
      <summary style='font-size:11px;opacity:0.6'>Detail analisis teknis</summary>
      <pre style='font-size:11px;margin:6px 0 0;opacity:0.7'>{}</pre>
    </details>
    </div>""".format(items, detail)


def adaptive_preprocess(gray, std_dev, lap_var):
    processed = gray.copy()
    steps_used = []

    if lap_var < 80:
        kernel_sharp = np.array([[0, -1, 0], [-1, 5, -1], [0, -1, 0]])
        processed = cv2.filter2D(processed, -1, kernel_sharp)
        steps_used.append("Sharpening (koreksi blur)")

    if std_dev < 40:
        clahe = cv2.createCLAHE(clipLimit=3.0, tileGridSize=(8, 8))
        processed = clahe.apply(processed)
        steps_used.append("CLAHE (perkuat kontras lokal)")

    processed = cv2.GaussianBlur(processed, (3, 3), 0)
    steps_used.append("Gaussian Blur 3×3 (haluskan noise)")

    return processed, steps_used


# ==========================================================
# PIPELINE UTAMA (Safe Concurrency)
# ==========================================================
def process_signature(image):
    # Disesuaikan murni menjadi 17 Output (Menghapus placeholder State rusak)
    NOUT = 17
    if image is None:
        return (None,) * NOUT

    # Menghindari penumpukan file antar user via folder temporer unik
    tmp_dir = tempfile.mkdtemp()

    img = cv2.cvtColor(image, cv2.COLOR_RGB2BGR)
    h0, w0 = img.shape[:2]

    # ── ANALISIS KUALITAS ─────────────────────────────────
    gray_raw = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    std_dev  = np.std(gray_raw)
    lap_var  = cv2.Laplacian(gray_raw, cv2.CV_64F).var()

    warnings, suggestions, is_ok, detail = analyze_image_quality(gray_raw, h0, w0)
    diagnosis_html = build_diagnosis_html(warnings, suggestions, is_ok, detail)

    # ── TAHAP 1: GRAYSCALE ────────────────────────────────
    gray = gray_raw.copy()
    gray_display = cv2.cvtColor(gray, cv2.COLOR_GRAY2RGB)
    info_gray = f"Resolusi  : {w0} × {h0} px\nChannel   : 1 (Grayscale)\nNilai px  : 0–255\nRata-rata : {np.mean(gray):.0f}\nStd dev   : {std_dev:.1f}"

    # ── TAHAP 2: PRE-PROCESSING ADAPTIF + THRESHOLD ───────
    preprocessed, steps_used = adaptive_preprocess(gray, std_dev, lap_var)
    otsu_val, thresh = cv2.threshold(preprocessed, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

    thresh_display = cv2.cvtColor(thresh, cv2.COLOR_GRAY2RGB)
    adaptasi_info  = "\n".join(["• " + s for s in steps_used]) if steps_used else "• Tidak diperlukan (gambar sudah baik)"
    info_thresh = f"Metode      : Otsu's Binarization\nThreshold   : {otsu_val:.0f} (otomatis)\nPre-process :\n{adaptasi_info}"

    # ── TAHAP 3: MORFOLOGI ────────────────────────────────
    inv = cv2.bitwise_not(thresh)
    k_open  = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (3, 3))
    k_close = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (3, 3))
    opened  = cv2.morphologyEx(inv, cv2.MORPH_OPEN,  k_open,  iterations=1)
    closed  = cv2.morphologyEx(opened, cv2.MORPH_CLOSE, k_close, iterations=2)

    morph_display = cv2.cvtColor(cv2.bitwise_not(closed), cv2.COLOR_GRAY2RGB)
    noise_before  = np.sum(inv    == 255)
    noise_after   = np.sum(closed == 255)
    info_morph = "Opening   → hapus noise/titik kecil\nClosing   → tutup celah di stroke\nKernel    → Ellipse 3×3\nPiksel aktif sebelum : {:,}\nPiksel aktif sesudah : {:,}\nNoise dihapus        : {:,} px".format(
        noise_before, noise_after, max(0, noise_before - noise_after))

    # ── TAHAP 4: DETEKSI KONTUR (Koreksi Logika Warna Ungu BGR) ──
    contours, _ = cv2.findContours(closed, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_TC89_KCOS)
    min_area      = 80
    valid_contours = [c for c in contours if cv2.contourArea(c) > min_area]

    contour_vis = img.copy()
    # Warna #7c6fcd diatur sebagai (205, 111, 124) karena OpenCV memakai skema BGR
    cv2.drawContours(contour_vis, valid_contours, -1, (205, 111, 124), 2)
    contour_display = cv2.cvtColor(contour_vis, cv2.COLOR_BGR2RGB)
    info_contour = "Total kontur      : {}\nKontur valid      : {} (area > {}px)\nKontur noise      : {} (dibuang)\nOverlay           : ungu (#7c6fcd)\nMetode approx     : TC89_KCOS".format(
        len(contours), len(valid_contours), min_area, len(contours)-len(valid_contours))

    # ── TAHAP 5: CROP & RESIZE 400×400 ────────────────────
    coords = cv2.findNonZero(closed)
    if coords is None:
        empty = np.ones((400, 400, 3), dtype=np.uint8) * 30
        err   = "Tidak ada objek terdeteksi.\nCoba upload ulang."
        return (diagnosis_html, gray_display, info_gray, thresh_display, info_thresh,
                morph_display, info_morph, empty, err, empty, err, None, err,
                None, None, None, None)

    x, y, w, h = cv2.boundingRect(coords)
    pad = 20
    x1, y1 = max(0, x - pad), max(0, y - pad)
    x2, y2 = min(closed.shape[1], x + w + pad), min(closed.shape[0], y + h + pad)
    cropped = closed[y1:y2, x1:x2]

    TARGET = 400
    scale  = min(TARGET / cropped.shape[1], TARGET / cropped.shape[0])
    nw, nh = int(cropped.shape[1] * scale), int(cropped.shape[0] * scale)
    resized = cv2.resize(cropped, (nw, nh), interpolation=cv2.INTER_AREA)
    _, resized = cv2.threshold(resized, 127, 255, cv2.THRESH_BINARY)

    canvas = np.zeros((TARGET, TARGET), dtype=np.uint8)
    ox, oy = (TARGET - nw) // 2, (TARGET - nh) // 2
    canvas[oy:oy+nh, ox:ox+nw] = resized

    final_png     = cv2.bitwise_not(canvas)
    final_display = cv2.cvtColor(final_png, cv2.COLOR_GRAY2RGB)
    info_resize   = "Bounding box : {}×{} px\nPadding      : {} px\nCrop area    : {}×{} px\nSkala resize : {:.3f}×\nKanvas akhir : {}×{} px\nOffset pusat : ({}, {})".format(
        w, h, pad, x2-x1, y2-y1, scale, TARGET, TARGET, ox, oy)

    # ── TAHAP 6: EXPORT SVG ───────────────────────────────
    path_svg = os.path.join(tmp_dir, "signature_vectorized.svg")
    dwg = svgwrite.Drawing(path_svg, size=("400px", "400px"), profile="full")
    svg_contours, _ = cv2.findContours(canvas, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_TC89_KCOS)
    path_count = 0
    for cnt in svg_contours:
        if cv2.contourArea(cnt) > 80:
            approx = cv2.approxPolyDP(cnt, epsilon=0.8, closed=True)
            pts    = [(int(p[0][0]), int(p[0][1])) for p in approx]
            if len(pts) >= 3:
                dwg.add(dwg.polygon(points=pts, fill="black", stroke="black", stroke_width=0.3))
                path_count += 1
    dwg.save()
    info_svg = f"Format     : SVG (Scalable Vector Graphics)\nPath/polygon: {path_count}\nViewbox     : 400×400\nSkalabilitas: ∞ (tidak pecah di-zoom)"

    # ── SIMPAN FILE ISOALSI SESI ───────────────────────────
    path_clean = os.path.join(tmp_dir, "signature_clean.png")
    cv2.imwrite(path_clean, final_png)

    rgba = cv2.cvtColor(final_png, cv2.COLOR_GRAY2BGRA)
    rgba[final_png == 255] = [255, 255, 255, 0]
    rgba[final_png == 0]   = [0, 0, 0, 255]
    path_transparent = os.path.join(tmp_dir, "signature_transparent.png")
    cv2.imwrite(path_transparent, rgba)

    path_csv = os.path.join(tmp_dir, "signature_matrix.csv")
    pd.DataFrame(canvas).to_csv(path_csv, index=False)

    path_zip = os.path.join(tmp_dir, "hasil_signature.zip")
    with zipfile.ZipFile(path_zip, "w") as zf:
        zf.write(path_clean, arcname="signature_clean.png")
        zf.write(path_transparent, arcname="signature_transparent.png")
        zf.write(path_svg, arcname="signature_vectorized.svg")
        zf.write(path_csv, arcname="signature_matrix.csv")

    return (
        diagnosis_html,
        gray_display,    info_gray,
        thresh_display,  info_thresh,
        morph_display,   info_morph,
        contour_display, info_contour,
        final_display,   info_resize,      # Mengirim berkas path terisolasi
        path_svg,        info_svg,
        path_clean,
        path_transparent,
        path_csv,
        path_zip,
    )


# ==========================================================
# UI
# ==========================================================
with gr.Blocks(css=CUSTOM_CSS, title="Signature to Vector") as demo:

    gr.HTML("""
    <div class="app-header">
        <h1>✍️ Konversi Tanda Tangan ke Format Vektor</h1>
        <p>Grayscale → Pre-processing Adaptif → Thresholding → Morfologi → Kontur → Resize → SVG</p>
    </div>
    """)

    with gr.Row(equal_height=False):

        # ── PANEL KIRI ──
        with gr.Column(scale=1, min_width=260):
            gr.HTML('<div class="card">')
            gr.HTML('<div class="step-badge">📥 Input</div>')
            gr.HTML('<div class="step-title">Upload Tanda Tangan</div>')
            gr.HTML('<div class="step-desc">Foto / scan tanda tangan. Bisa pulpen maupun pensil — sistem akan menyesuaikan otomatis.</div>')
            input_image = gr.Image(type="numpy", label="Citra Tanda Tangan")
            btn = gr.Button("▶  Jalankan Pipeline", variant="primary")
            gr.HTML('</div>')

            gr.HTML('<div class="card">')
            gr.HTML('<div class="step-badge">📦 Unduh Hasil</div>')
            out_zip  = gr.File(label="📁 Paket Lengkap (.zip)")
            out_png  = gr.File(label="🖼  PNG Bersih")
            out_trns = gr.File(label="🫧  PNG Transparan")
            out_csv  = gr.File(label="📊 Matriks (.csv)")
            gr.HTML('</div>')

        # ── PANEL KANAN ──
        with gr.Column(scale=3):

            # DIAGNOSIS
            gr.HTML("""
            <div class="card">
              <div class="step-badge">🔍 Diagnosis Awal</div>
              <div class="step-title">Analisis Kualitas Gambar</div>
              <div class="step-desc">Sebelum diproses, sistem menganalisis kondisi foto: kecerahan, kontras, ketajaman, dan kebersihan background. Jika ada masalah, sistem memberikan saran dan menyesuaikan pipeline secara otomatis.</div>
            </div>
            """)
            diagnosis_out = gr.HTML(value="<div class='warn-box' style='opacity:0.4'>Hasil diagnosis akan muncul di sini setelah gambar diproses...</div>")
            gr.HTML('<div class="flow-arrow">↓</div>')

            # TAHAP 1
            gr.HTML("""
            <div class="card">
              <div class="step-badge">Tahap 1</div>
              <div class="step-title">Grayscale — Konversi ke Citra Keabuan</div>
              <div class="step-desc">Citra RGB 3-channel diubah ke 1-channel (0–255). Menyederhanakan data agar threshold lebih akurat.</div>
            </div>
            """)
            with gr.Row():
                img_gray  = gr.Image(label="Hasil Grayscale", type="numpy", height=220)
                info_gray = gr.Textbox(label="ℹ️ Informasi", lines=6, interactive=False)
            gr.HTML('<div class="flow-arrow">↓</div>')

            # TAHAP 2
            gr.HTML("""
            <div class="card">
              <div class="step-badge">Tahap 2</div>
              <div class="step-title">Pre-processing Adaptif + Thresholding Otsu</div>
              <div class="step-desc">Sistem mendeteksi kondisi gambar: jika buram → ditajamkan, jika kontras rendah (pensil) → CLAHE aktif. Setelah itu Otsu menentukan threshold binarisasi secara otomatis.</div>
            </div>
            """)
            with gr.Row():
                img_thresh  = gr.Image(label="Hasil Binarisasi", type="numpy", height=220)
                info_thresh = gr.Textbox(label="ℹ️ Informasi", lines=6, interactive=False)
            gr.HTML('<div class="flow-arrow">↓</div>')

            # TAHAP 3
            gr.HTML("""
            <div class="card">
              <div class="step-badge">Tahap 3</div>
              <div class="step-title">Morfologi — Bersihkan Noise & Perkuat Stroke</div>
              <div class="step-desc">Opening menghapus titik noise kecil sisa threshold. Closing menutup celah kecil dalam garis agar stroke tanda tangan solid dan rapi.</div>
            </div>
            """)
            with gr.Row():
                img_morph  = gr.Image(label="Setelah Morfologi", type="numpy", height=220)
                info_morph = gr.Textbox(label="ℹ️ Informasi", lines=6, interactive=False)
            gr.HTML('<div class="flow-arrow">↓</div>')

            # TAHAP 4
            gr.HTML("""
            <div class="card">
              <div class="step-badge">Tahap 4</div>
              <div class="step-title">Deteksi Kontur — Temukan Batas Stroke</div>
              <div class="step-desc">findContours menelusuri batas luar setiap stroke. Kontur kecil (noise) difilter. Kontur valid ditampilkan sebagai overlay ungu di atas citra asli.</div>
            </div>
            """)
            with gr.Row():
                img_contour  = gr.Image(label="Overlay Kontur", type="numpy", height=220)
                info_contour = gr.Textbox(label="ℹ️ Informasi", lines=6, interactive=False)
            gr.HTML('<div class="flow-arrow">↓</div>')

            # TAHAP 5
            gr.HTML("""
            <div class="card">
              <div class="step-badge">Tahap 5</div>
              <div class="step-title">Crop & Resize — Normalisasi ke 400×400</div>
              <div class="step-desc">Bounding box mendeteksi area aktif, lalu di-crop dan di-resize ke kanvas standar 400×400 px. Aspect ratio tetap terjaga, objek diposisikan di tengah.</div>
            </div>
            """)
            with gr.Row():
                img_final  = gr.Image(label="Hasil Akhir 400×400", type="numpy", height=220)
                info_final = gr.Textbox(label="ℹ️ Informasi", lines=6, interactive=False)
            gr.HTML('<div class="flow-arrow">↓</div>')

            # TAHAP 6
            gr.HTML("""
            <div class="card">
              <div class="step-badge">Tahap 6</div>
              <div class="step-title">Vektorisasi SVG — Output Format Vektor</div>
              <div class="step-desc">Kontur dikonversi menjadi polygon SVG. Format vektor dapat di-zoom tak terbatas tanpa pecah — ideal untuk tanda tangan digital.</div>
            </div>
            """)
            with gr.Row():
                # Diubah ke komponen gr.File karena gr.Image tidak stabil menampilkan file .svg murni di beberapa peramban web
                img_svg  = gr.File(label="Berkas Vektor SVG Hasil Ekstraksi")
                info_svg = gr.Textbox(label="ℹ️ Informasi", lines=6, interactive=False)

    # 17 Inputs and Outputs matched perfectly
    btn.click(
        fn=process_signature,
        inputs=[input_image],
        outputs=[
            diagnosis_out,
            img_gray,    info_gray,
            img_thresh,  info_thresh,
            img_morph,   info_morph,
            img_contour, info_contour,
            img_final,   info_final,
            img_svg,     info_svg,
            out_png, out_trns, out_csv, out_zip
        ]
    )

demo.launch(share=True)
```

```python
import cv2
import gradio as gr
import numpy as np
import svgwrite
import zipfile
import pandas as pd
import io
import os
import tempfile

from skimage import morphology

# ==========================================================
# CSS DARK MODE ELEGAN & CHECKERBOARD BACKGROUND
# ==========================================================
CUSTOM_CSS = """
:root {
    --bg-main:    #0f1117;
    --bg-card:    #1a1d27;
    --bg-input:   #12151f;
    --border:     #2a2d3a;
    --accent:     #7c6fcd;
    --accent-dim: #3d3670;
    --text-main:  #e8e8f0;
    --text-muted: #7a7a9a;
    --text-label: #a0a0c0;
    --radius:     12px;
}
body, .gradio-container {
    background-color: var(--bg-main) !important;
    color: var(--text-main) !important;
    font-family: 'Inter', 'Segoe UI', sans-serif !important;
}
.app-header {
    text-align: center;
    padding: 36px 20px 24px;
    border-bottom: 1px solid var(--border);
    margin-bottom: 24px;
}
.app-header h1 { font-size: 26px; font-weight: 700; color: var(--text-main); margin: 0 0 6px; }
.app-header p  { color: var(--text-muted); font-size: 13px; margin: 0; }

.card {
    background: var(--bg-card);
    border: 1px solid var(--border);
    border-radius: var(--radius);
    padding: 18px 20px;
    margin-bottom: 14px;
}
.step-badge {
    display: inline-flex;
    align-items: center;
    gap: 6px;
    background: var(--accent-dim);
    color: #c4b8ff;
    font-size: 11px;
    font-weight: 600;
    letter-spacing: 0.8px;
    text-transform: uppercase;
    padding: 3px 10px;
    border-radius: 20px;
    margin-bottom: 5px;
}
.step-title { font-size: 15px; font-weight: 600; color: var(--text-main); margin: 0 0 3px; }
.step-desc  { font-size: 12px; color: var(--text-muted); margin: 0; line-height: 1.5; }

.warn-box {
    background: #2a1f0e;
    border: 1px solid #7a4f1a;
    border-radius: 8px;
    padding: 12px 16px;
    font-size: 13px;
    color: #f5c97a;
    margin-bottom: 14px;
    line-height: 1.6;
}
.ok-box {
    background: #0e2a1a;
    border: 1px solid #1a6640;
    border-radius: 8px;
    padding: 12px 16px;
    font-size: 13px;
    color: #6ee7a8;
    margin-bottom: 14px;
    line-height: 1.6;
}
.pipeline-box {
    background: #0e1a2a;
    border: 1px solid #1a4f7a;
    border-radius: 8px;
    padding: 12px 16px;
    font-size: 13px;
    color: #7ac4f5;
    margin-bottom: 14px;
    line-height: 1.6;
}

.gr-button-primary {
    background: var(--accent) !important;
    border: none !important;
    color: #fff !important;
    border-radius: 8px !important;
    font-weight: 600 !important;
    padding: 12px 0 !important;
    transition: opacity 0.2s !important;
}
.gr-button-primary:hover { opacity: 0.85 !important; }

.gr-textbox textarea, .gr-textbox input {
    background: var(--bg-input) !important;
    border: 1px solid var(--border) !important;
    color: var(--text-main) !important;
    border-radius: 8px !important;
    font-family: 'JetBrains Mono', 'Courier New', monospace !important;
    font-size: 12px !important;
}
label span { color: var(--text-label) !important; font-size: 12px !important; font-weight: 500 !important; }
.gr-image  { border-radius: 8px !important; border: 1px solid var(--border) !important; background: #000 !important; }
.gr-file   { background: var(--bg-input) !important; border: 1px solid var(--border) !important; border-radius: 8px !important; }
.flow-arrow { text-align: center; color: var(--accent); font-size: 20px; margin: 2px 0; }

/* Background pattern untuk preview transparan */
.transparent-preview { background-color: #f0f0f0 !important; background-image: linear-gradient(45deg, #d4d4d4 25%, transparent 25%), linear-gradient(-45deg, #d4d4d4 25%, transparent 25%), linear-gradient(45deg, transparent 75%, #d4d4d4 75%), linear-gradient(-45deg, transparent 75%, #d4d4d4 75%) !important; background-size: 20px 20px !important; background-position: 0 0, 0 10px, 10px -10px, -10px 0px !important; }
"""

# ==========================================================
# HELPER: KONVERSI ARRAY KE DATAFRAME
# ==========================================================
def arr_to_df(arr):
    """Mengubah array numpy (2D atau 3D) menjadi Pandas DataFrame untuk ditampilkan di gr.DataFrame"""
    if arr is None:
        return pd.DataFrame()
    if arr.ndim == 2:
        return pd.DataFrame(arr)
    elif arr.ndim == 3:
        # Jika RGB/RGBA, gabungkan nilai channel menjadi string "R,G,B" per sel agar muat di satu tabel
        arr_str = arr[:, :, 0].astype(str)
        for c in range(1, arr.shape[2]):
            arr_str = np.char.add(arr_str, ",")
            arr_str = np.char.add(arr_str, arr[:, :, c].astype(str))
        return pd.DataFrame(arr_str)
    return pd.DataFrame()

# ==========================================================
# ANALISIS KUALITAS GAMBAR
# ==========================================================
def analyze_image_quality(gray, h, w):
    warnings = []
    suggestions = []

    mean_bright = np.mean(gray)
    if mean_bright < 80:
        warnings.append("⚠️  Gambar terlalu gelap (rata-rata kecerahan: {:.0f}/255)".format(mean_bright))
        suggestions.append("→ Foto ulang di tempat yang lebih terang, atau dekat lampu.")
    elif mean_bright > 230:
        warnings.append("⚠️  Gambar terlalu terang/overexposed (rata-rata: {:.0f}/255)".format(mean_bright))
        suggestions.append("→ Kurangi cahaya atau jangan terlalu dekat sumber cahaya.")

    std_dev = np.std(gray)
    if std_dev < 20:
        warnings.append("⚠️  Kontras sangat rendah (std: {:.1f}) — kemungkinan tanda tangan pensil tipis atau foto buram.".format(std_dev))
        suggestions.append("→ Gunakan pensil/pulpen yang lebih tebal, atau perbesar kontras saat foto.")

    lap_var = cv2.Laplacian(gray, cv2.CV_64F).var()
    if lap_var < 50:
        warnings.append("⚠️  Gambar buram / tidak fokus (sharpness score: {:.1f})".format(lap_var))
        suggestions.append("→ Foto dengan kamera lebih stabil, pastikan tanda tangan dalam fokus.")

    _, tmp = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    dark_ratio = np.sum(tmp == 0) / (h * w)
    if dark_ratio > 0.5:
        warnings.append("⚠️  Lebih dari 50% piksel terdeteksi gelap ({:.0f}%) — kemungkinan ada bayangan besar atau background kotor.".format(dark_ratio*100))
        suggestions.append("→ Foto di kertas putih bersih, hindari bayangan tangan/objek lain.")

    if w < 200 or h < 200:
        warnings.append("⚠️  Resolusi terlalu kecil ({}×{} px) — hasil vektorisasi bisa kasar.".format(w, h))
        suggestions.append("→ Upload foto dengan resolusi minimal 400×400 px.")

    detail = "Kecerahan rata-rata : {:.0f}/255\nKontras (std dev)   : {:.1f}\nSharpness score     : {:.1f}\nArea gelap          : {:.1f}%\nResolusi            : {}×{} px".format(
        mean_bright, std_dev, lap_var, dark_ratio*100, w, h
    )

    is_ok = len(warnings) == 0
    return warnings, suggestions, is_ok, detail


def build_diagnosis_html(warnings, suggestions, is_ok, detail, pipeline_steps):
    pipeline_html = ""
    if pipeline_steps:
        items = "".join("<li style='margin-bottom:4px'>{}</li>".format(s) for s in pipeline_steps)
        pipeline_html = """<div class="pipeline-box">
        <strong>🔧 Penyesuaian Pipeline Otomatis:</strong>
        <ul style='margin:8px 0 0 16px;padding:0'>{}</ul>
        </div>""".format(items)

    if is_ok:
        base = """<div class="ok-box">✅ <strong>Gambar terdeteksi baik!</strong> Tidak ada masalah kualitas yang ditemukan.<br><br><small style='opacity:0.7'>{}</small></div>""".format(detail.replace('\n', '<br>'))
        return base + pipeline_html

    items = ""
    for w, s in zip(warnings, suggestions):
        items += "<li style='margin-bottom:6px'>{}<br><span style='color:#a0c4a0'>{}</span></li>".format(w, s)

    warn_html = """<div class="warn-box">
    <strong>⚠️ Terdeteksi masalah pada gambar — pipeline tetap dijalankan dengan penyesuaian otomatis:</strong>
    <ul style='margin:8px 0 0 16px;padding:0'>{}</ul>
    <details style='margin-top:10px;cursor:pointer'>
      <summary style='font-size:11px;opacity:0.6'>Detail analisis teknis</summary>
      <pre style='font-size:11px;margin:6px 0 0;opacity:0.7'>{}</pre>
    </details>
    </div>""".format(items, detail)

    return warn_html + pipeline_html


def adaptive_preprocess(gray, std_dev, lap_var):
    processed = gray.copy()
    steps_used = []
    pipeline_flags = []

    if lap_var < 80:
        kernel_sharp = np.array([[0, -1, 0], [-1, 5, -1], [0, -1, 0]])
        processed = cv2.filter2D(processed, -1, kernel_sharp)
        steps_used.append("Sharpening (koreksi blur)")

    clahe_active = std_dev < 45
    if clahe_active:
        clahe = cv2.createCLAHE(clipLimit=3.0, tileGridSize=(8, 8))
        processed = clahe.apply(processed)
        steps_used.append("CLAHE (perkuat kontras lokal)")
        pipeline_flags.append("✓ CLAHE aktif — kontras lokal diperkuat untuk garis pensil/tipis")

    processed = cv2.bilateralFilter(processed, d=7, sigmaColor=50, sigmaSpace=50)
    steps_used.append("Bilateral Filter (redam noise, jaga ketajaman tepi)")
    pipeline_flags.append("✓ Bilateral Filter aktif — noise berkurang, tepi garis tetap tajam")

    return processed, steps_used, pipeline_flags


def smart_threshold(preprocessed, h0, w0):
    pipeline_flags = []
    total_px = h0 * w0

    otsu_val, thresh_otsu = cv2.threshold(
        preprocessed, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU
    )
    active_ratio_otsu = np.sum(thresh_otsu == 0) / total_px  # Koreksi perhitungan objek biner gelap

    otsu_bad = (active_ratio_otsu < 0.002) or (active_ratio_otsu > 0.35)

    if not otsu_bad:
        pipeline_flags.append("✓ Otsu Threshold digunakan — hasil binarisasi dinilai baik")
        return thresh_otsu, "Otsu", otsu_val, pipeline_flags

    block_size = max(11, (min(h0, w0) // 20) | 1)
    thresh_adaptive = cv2.adaptiveThreshold(
        preprocessed, 255,
        cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
        cv2.THRESH_BINARY,
        block_size,
        5
    )

    pipeline_flags.append(
        "✓ Adaptive Threshold aktif (fallback dari Otsu — rasio piksel aktif Otsu {:.2f}%)".format(active_ratio_otsu * 100)
    )
    return thresh_adaptive, "Adaptive Gaussian", otsu_val, pipeline_flags


# ==========================================================
# PIPELINE UTAMA (Safe Concurrency)
# ==========================================================
def process_signature(image):
    # Total Output = 25
    NOUT = 25
    if image is None:
        return (None,) * NOUT

    tmp_dir = tempfile.mkdtemp()

    img = cv2.cvtColor(image, cv2.COLOR_RGB2BGR)
    h0, w0 = img.shape[:2]

    # ── ANALISIS KUALITAS ─────────────────────────────────
    gray_raw = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    std_dev  = np.std(gray_raw)
    lap_var  = cv2.Laplacian(gray_raw, cv2.CV_64F).var()

    warnings, suggestions, is_ok, detail = analyze_image_quality(gray_raw, h0, w0)

    # ── TAHAP 1: GRAYSCALE ────────────────────────────────
    gray = gray_raw.copy()
    gray_display = cv2.cvtColor(gray, cv2.COLOR_GRAY2RGB)
    df_gray = arr_to_df(gray) # Matriks ukuran asli
    info_gray = f"Resolusi  : {w0} × {h0} px\nChannel   : 1 (Grayscale)\nNilai px  : 0–255\nRata-rata : {np.mean(gray):.0f}\nStd dev   : {std_dev:.1f}"

    # ── TAHAP 2: PRE-PROCESSING ADAPTIF + THRESHOLD ───
    preprocessed, steps_used, pre_flags = adaptive_preprocess(gray, std_dev, lap_var)
    thresh, method_name, otsu_val, thresh_flags = smart_threshold(preprocessed, h0, w0)

    pipeline_flags = pre_flags + thresh_flags
    diagnosis_html = build_diagnosis_html(warnings, suggestions, is_ok, detail, pipeline_flags)

    thresh_display = cv2.cvtColor(thresh, cv2.COLOR_GRAY2RGB)
    df_thresh = arr_to_df(thresh) # Matriks ukuran asli
    adaptasi_info  = "\n".join(["• " + s for s in steps_used]) if steps_used else "• Tidak diperlukan (gambar sudah baik)"
    info_thresh = "Metode      : {}\nOtsu (ref)  : {:.0f}\nPre-process :\n{}".format(
        method_name, otsu_val, adaptasi_info
    )

    # ── TAHAP 3: MORFOLOGI ────────────────────────────────
    inv = cv2.bitwise_not(thresh)
    k_open  = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (3, 3))
    opened  = cv2.morphologyEx(inv, cv2.MORPH_OPEN, k_open, iterations=1)
    closed = opened.copy()

    morph_display = cv2.cvtColor(cv2.bitwise_not(closed), cv2.COLOR_GRAY2RGB)
    df_morph = arr_to_df(closed) # Matriks ukuran asli
    noise_before  = np.sum(inv    == 255)
    noise_after   = np.sum(closed == 255)
    info_morph = "Opening   → hapus noise/titik kecil sisa binarisasi\nClosing   → dinonaktifkan sesuai evaluasi (menjaga detail stroke)\nKernel    → Ellipse 3×3\nPiksel aktif sebelum : {:,}\nPiksel aktif sesudah : {:,}\nNoise dihapus        : {:,} px".format(
        noise_before, noise_after, max(0, noise_before - noise_after))

    # ── TAHAP 4: DETEKSI KONTUR (Area Adaptif) ────────
    contours, hierarchy = cv2.findContours(closed, cv2.RETR_TREE, cv2.CHAIN_APPROX_TC89_KCOS)

    total_px = h0 * w0
    min_area = max(6, total_px * 0.00005)
    valid_contours = [c for c in contours if cv2.contourArea(c) > min_area]

    contour_vis = img.copy()
    cv2.drawContours(contour_vis, valid_contours, -1, (205, 111, 124), 2) # Menggambar garis ungu
    contour_display = cv2.cvtColor(contour_vis, cv2.COLOR_BGR2RGB)
    df_contour = arr_to_df(contour_display) # Matriks RGB ukuran asli format "R,G,B"
    info_contour = "Total kontur      : {}\nKontur valid      : {} (area > {:.1f}px, adaptif)\nKontur noise      : {} (dibuang)\nOverlay           : ungu (#7c6fcd)\nMetode approx     : TC89_KCOS".format(
        len(contours), len(valid_contours), min_area, len(contours)-len(valid_contours))

    # ── TAHAP 5: CROP & RESIZE 400×400 ────────────────────
    coords = cv2.findNonZero(closed)
    if coords is None:
        empty = np.ones((400, 400, 3), dtype=np.uint8) * 30
        empty_df = pd.DataFrame()
        err   = "Tidak ada objek terdeteksi.\nCoba upload ulang."
        return (diagnosis_html,
                gray_display, info_gray, df_gray,
                thresh_display, info_thresh, df_thresh,
                morph_display, info_morph, df_morph,
                contour_display, info_contour, df_contour,
                empty, err, empty_df,
                empty, err, empty_df,
                None, err,
                None, None, None, None)

    x, y, w, h = cv2.boundingRect(coords)
    pad = 20
    x1, y1 = max(0, x - pad), max(0, y - pad)
    x2, y2 = min(closed.shape[1], x + w + pad), min(closed.shape[0], y + h + pad)
    cropped = closed[y1:y2, x1:x2]

    TARGET = 400
    scale  = min(TARGET / cropped.shape[1], TARGET / cropped.shape[0])
    nw, nh = int(cropped.shape[1] * scale), int(cropped.shape[0] * scale)
    resized = cv2.resize(cropped, (nw, nh), interpolation=cv2.INTER_AREA)
    _, resized = cv2.threshold(resized, 127, 255, cv2.THRESH_BINARY)

    canvas = np.zeros((TARGET, TARGET), dtype=np.uint8)
    ox, oy = (TARGET - nw) // 2, (TARGET - nh) // 2
    canvas[oy:oy+nh, ox:ox+nw] = resized

    final_png     = cv2.bitwise_not(canvas)
    final_display = cv2.cvtColor(final_png, cv2.COLOR_GRAY2RGB)
    df_final      = arr_to_df(final_png) # Matriks 400x400 (karena ini stage crop & resize)
    info_resize   = "Bounding box : {}×{} px\nPadding      : {} px\nCrop area    : {}×{} px\nSkala resize : {:.3f}×\nKanvas akhir : {}×{} px\nOffset pusat : ({}, {})".format(
        w, h, pad, x2-x1, y2-y1, scale, TARGET, TARGET, ox, oy)

    # ── TAHAP 5.5: PREVIEW BACKGROUND REMOVAL ──
    rgba = cv2.cvtColor(final_png, cv2.COLOR_GRAY2BGRA)
    rgba[final_png == 255] = [255, 255, 255, 0]
    rgba[final_png == 0]   = [0, 0, 0, 255]

    df_transparent = arr_to_df(rgba) # Matriks RGBA 400x400 format "R,G,B,A"

    # Komposit catur untuk preview gradionya (Display saja)
    checker = np.zeros((TARGET, TARGET, 3), dtype=np.uint8)
    cell = 20
    for cy in range(0, TARGET, cell):
        for cx in range(0, TARGET, cell):
            color = 240 if ((cx // cell) + (cy // cell)) % 2 == 0 else 215
            checker[cy:cy+cell, cx:cx+cell] = (color, color, color)

    alpha = rgba[:, :, 3:4].astype(np.float32) / 255.0
    rgb_fg = rgba[:, :, :3].astype(np.float32)
    transparent_preview = (rgb_fg * alpha + checker.astype(np.float32) * (1 - alpha)).astype(np.uint8)

    transparent_pixel_count = int(np.sum(final_png == 0))
    info_transparent = "Status        : Latar belakang dibersihkan (saluran Alpha)\nArea objek    : {:,} px (tanda tangan)\nArea kosong   : {:,} px (transparan)\nVisualisasi   : Matriks komposit grid papan catur\nDimensi data  : {}×{} px".format(
        transparent_pixel_count, TARGET*TARGET - transparent_pixel_count, TARGET, TARGET
    )

    # ── TAHAP 6: EXPORT SVG ──
    path_svg = os.path.join(tmp_dir, "signature_vectorized.svg")
    dwg = svgwrite.Drawing(path_svg, size=("400px", "400px"), profile="full")

    svg_contours, svg_hier = cv2.findContours(canvas, cv2.RETR_TREE, cv2.CHAIN_APPROX_TC89_KCOS)
    path_count = 0

    if svg_hier is not None:
        for idx, cnt in enumerate(svg_contours):
            if cv2.contourArea(cnt) > min_area:
                approx = cv2.approxPolyDP(cnt, epsilon=0.8, closed=True)
                pts    = [(int(p[0][0]), int(p[0][1])) for p in approx]
                if len(pts) >= 3:
                    dwg.add(dwg.polygon(points=pts, fill="black", stroke="black", stroke_width=0.3))
                    path_count += 1

    dwg.save()
    info_svg = f"Format     : SVG (Scalable Vector Graphics)\nPath/polygon: {path_count}\nViewbox    : 400×400\nAturan Isi : Hierarchy Tree (lubang tengah aman)"

    # ── SIMPAN FILE ISOLASI SESI ───────────────────────────
    path_clean = os.path.join(tmp_dir, "signature_clean.png")
    cv2.imwrite(path_clean, final_png)

    path_transparent = os.path.join(tmp_dir, "signature_transparent.png")
    cv2.imwrite(path_transparent, rgba)

    path_csv = os.path.join(tmp_dir, "signature_matrix.csv")
    pd.DataFrame(canvas).to_csv(path_csv, index=False)

    path_zip = os.path.join(tmp_dir, "hasil_signature.zip")
    with zipfile.ZipFile(path_zip, "w") as zf:
        zf.write(path_clean, arcname="signature_clean.png")
        zf.write(path_transparent, arcname="signature_transparent.png")
        zf.write(path_svg, arcname="signature_vectorized.svg")
        zf.write(path_csv, arcname="signature_matrix.csv")

    return (
        diagnosis_html,
        gray_display, info_gray, df_gray,
        thresh_display, info_thresh, df_thresh,
        morph_display, info_morph, df_morph,
        contour_display, info_contour, df_contour,
        final_display, info_resize, df_final,
        transparent_preview, info_transparent, df_transparent,
        path_svg, info_svg,
        path_clean, path_transparent, path_csv, path_zip,
    )


# ==========================================================
# UI
# ==========================================================
with gr.Blocks(css=CUSTOM_CSS, title="Signature to Vector") as demo:

    gr.HTML("""
    <div class="app-header">
        <h1>✍️ Konversi Tanda Tangan ke Format Vektor</h1>
        <p>Grayscale → Pre-processing Adaptif → Thresholding → Morfologi → Kontur → Resize → Background Removal → SVG</p>
    </div>
    """)

    with gr.Row(equal_height=False):

        # ── PANEL KIRI ──
        with gr.Column(scale=1, min_width=260):
            gr.HTML('<div class="card">')
            gr.HTML('<div class="step-badge">📥 Input</div>')
            gr.HTML('<div class="step-title">Upload Tanda Tangan</div>')
            gr.HTML('<div class="step-desc">Foto / scan tanda tangan. Bisa pulpen maupun pensil — sistem akan menyesuaikan otomatis.</div>')
            input_image = gr.Image(type="numpy", label="Citra Tanda Tangan")
            btn = gr.Button("▶  Jalankan Pipeline", variant="primary")
            gr.HTML('</div>')

            gr.HTML('<div class="card">')
            gr.HTML('<div class="step-badge">📦 Unduh Hasil</div>')
            out_zip  = gr.File(label="📁 Paket Lengkap (.zip)")
            out_png  = gr.File(label="🖼  PNG Bersih")
            out_trns = gr.File(label="🫧  PNG Transparan")
            out_csv  = gr.File(label="📊 Matriks (.csv)")
            gr.HTML('</div>')

        # ── PANEL KANAN ──
        with gr.Column(scale=3):

            # DIAGNOSIS
            gr.HTML("""
            <div class="card">
              <div class="step-badge">🔍 Diagnosis Awal</div>
              <div class="step-title">Analisis Kualitas Gambar</div>
              <div class="step-desc">Sistem mendiagnosis gambar, menampilkan informasi Pipeline dinamis seperti aktivasi CLAHE, Bilateral Filter, atau Adaptive Threshold.</div>
            </div>
            """)
            diagnosis_out = gr.HTML(value="<div class='warn-box' style='opacity:0.4'>Hasil diagnosis akan muncul di sini setelah gambar diproses...</div>")
            gr.HTML('<div class="flow-arrow">↓</div>')

            # TAHAP 1
            gr.HTML("""
            <div class="card">
              <div class="step-badge">Tahap 1</div>
              <div class="step-title">Grayscale — Konversi ke Citra Keabuan</div>
              <div class="step-desc">Citra RGB 3-channel diubah ke 1-channel (0–255). Menyederhanakan data agar threshold lebih akurat.</div>
            </div>
            """)
            with gr.Row():
                with gr.Column(scale=3):
                    with gr.Tabs():
                        with gr.TabItem("🖼️ Hasil Gambar"):
                            img_gray  = gr.Image(label="Hasil Grayscale", type="numpy", height=280)
                        with gr.TabItem("🔢 Matrix Pixel"):
                            df_gray = gr.DataFrame(label="Matriks Grayscale", interactive=False)
                with gr.Column(scale=1):
                    info_gray = gr.Textbox(label="ℹ️ Informasi", lines=8, interactive=False)
            gr.HTML('<div class="flow-arrow">↓</div>')

            # TAHAP 2
            gr.HTML("""
            <div class="card">
              <div class="step-badge">Tahap 2</div>
              <div class="step-title">Pre-processing Adaptif + Thresholding</div>
              <div class="step-desc">Thresholding cerdas yang dioptimalkan dengan CLAHE/Bilateral Filter menyesuaikan kualitas pencahayaan foto asli.</div>
            </div>
            """)
            with gr.Row():
                with gr.Column(scale=3):
                    with gr.Tabs():
                        with gr.TabItem("🖼️ Hasil Gambar"):
                            img_thresh  = gr.Image(label="Hasil Binarisasi", type="numpy", height=280)
                        with gr.TabItem("🔢 Matrix Pixel"):
                            df_thresh = gr.DataFrame(label="Matriks Biner", interactive=False)
                with gr.Column(scale=1):
                    info_thresh = gr.Textbox(label="ℹ️ Informasi", lines=8, interactive=False)
            gr.HTML('<div class="flow-arrow">↓</div>')

            # TAHAP 3
            gr.HTML("""
            <div class="card">
              <div class="step-badge">Tahap 3</div>
              <div class="step-title">Morfologi — Bersihkan Noise & Jaga Goresan</div>
              <div class="step-desc">Opening menghapus titik noise kecil sisa threshold. Closing dinonaktifkan untuk menjaga struktur garis tipis.</div>
            </div>
            """)
            with gr.Row():
                with gr.Column(scale=3):
                    with gr.Tabs():
                        with gr.TabItem("🖼️ Hasil Gambar"):
                            img_morph  = gr.Image(label="Setelah Morfologi", type="numpy", height=280)
                        with gr.TabItem("🔢 Matrix Pixel"):
                            df_morph = gr.DataFrame(label="Matriks Morfologi", interactive=False)
                with gr.Column(scale=1):
                    info_morph = gr.Textbox(label="ℹ️ Informasi", lines=8, interactive=False)
            gr.HTML('<div class="flow-arrow">↓</div>')

            # TAHAP 4
            gr.HTML("""
            <div class="card">
              <div class="step-badge">Tahap 4</div>
              <div class="step-title">Deteksi Kontur — Temukan Batas Stroke</div>
              <div class="step-desc">Tanda tangan dilacak dengan area ambang batas adaptif yang dinamis menyesuaikan besaran resolusi.</div>
            </div>
            """)
            with gr.Row():
                with gr.Column(scale=3):
                    with gr.Tabs():
                        with gr.TabItem("🖼️ Hasil Gambar"):
                            img_contour  = gr.Image(label="Overlay Kontur", type="numpy", height=280)
                        with gr.TabItem("🔢 Matrix Pixel (RGB)"):
                            df_contour = gr.DataFrame(label="Matriks Warna RGB Asli", interactive=False)
                with gr.Column(scale=1):
                    info_contour = gr.Textbox(label="ℹ️ Informasi", lines=8, interactive=False)
            gr.HTML('<div class="flow-arrow">↓</div>')

            # TAHAP 5
            gr.HTML("""
            <div class="card">
              <div class="step-badge">Tahap 5</div>
              <div class="step-title">Crop & Resize — Normalisasi ke 400×400</div>
              <div class="step-desc">Bounding box mendeteksi area aktif, lalu memotong dan menyesuaikannya menjadi bentuk akhir siap pakai.</div>
            </div>
            """)
            with gr.Row():
                with gr.Column(scale=3):
                    with gr.Tabs():
                        with gr.TabItem("🖼️ Hasil Gambar"):
                            img_final  = gr.Image(label="Hasil Akhir 400×400", type="numpy", height=280)
                        with gr.TabItem("🔢 Matrix Pixel"):
                            df_final = gr.DataFrame(label="Matriks 400x400", interactive=False)
                with gr.Column(scale=1):
                    info_final = gr.Textbox(label="ℹ️ Informasi", lines=8, interactive=False)
            gr.HTML('<div class="flow-arrow">↓</div>')

            # TAHAP 5.5: PREVIEW BACKGROUND REMOVAL
            gr.HTML("""
            <div class="card">
              <div class="step-badge">Tahap 5.5</div>
              <div class="step-title">Preview Background Removal</div>
              <div class="step-desc">Pratinjau hasil akhir format RGBA (transparan). Piksel background tembus pandang ditunjukkan oleh nilai Alpha=0.</div>
            </div>
            """)
            with gr.Row():
                with gr.Column(scale=3):
                    with gr.Tabs():with gr.TabItem("🖼️ Hasil Gambar"):
                            img_transparent  = gr.Image(label="Preview PNG Transparan", type="numpy", height=280, elem_classes="transparent-preview")
                        with gr.TabItem("🔢 Matrix Pixel (RGBA)"):
                            df_transparent = gr.DataFrame(label="Matriks Transparansi (RGBA)", interactive=False)
                with gr.Column(scale=1):
                    info_transparent = gr.Textbox(label="ℹ️ Informasi", lines=8, interactive=False)
            gr.HTML('<div class="flow-arrow">↓</div>')

            # TAHAP 6
            gr.HTML("""
            <div class="card">
              <div class="step-badge">Tahap 6</div>
              <div class="step-title">Vektorisasi SVG — Output Format Vektor</div>
              <div class="step-desc">Kontur dikonversi menjadi polygon SVG. Format vektor dapat di-zoom tak terbatas tanpa pecah — ideal untuk tanda tangan digital.</div>
            </div>
            """)
            with gr.Row():
                img_svg  = gr.File(label="Berkas Vektor SVG Hasil Ekstraksi")
                info_svg = gr.Textbox(label="ℹ️ Informasi", lines=6, interactive=False)

    # 25 Total Inputs and Outputs matched
    btn.click(
        fn=process_signature,
        inputs=[input_image],
        outputs=[
            diagnosis_out,
            img_gray, info_gray, df_gray,
            img_thresh, info_thresh, df_thresh,
            img_morph, info_morph, df_morph,
            img_contour, info_contour, df_contour,
            img_final, info_final, df_final,
            img_transparent, info_transparent, df_transparent,
            img_svg, info_svg,
            out_png, out_trns, out_csv, out_zip
        ]
    )

demo.launch(share=True)
```

```python
# ==========================================================
# JALANKAN BARIS INI DI SEL PERTAMA GOOGLE COLAB
# !pip install opencv-python-headless scikit-image matplotlib svgwrite gradio pandas -q
# ==========================================================

import cv2
import gradio as gr
import numpy as np
import svgwrite
import zipfile
import pandas as pd
import io
import os
import tempfile

from skimage import morphology
from matplotlib.figure import Figure
from matplotlib.backends.backend_agg import FigureCanvasAgg

# ==========================================================
# CSS DARK MODE — MODERN, BERSIH, FOKUS PADA ALUR PROSES
# ==========================================================
CUSTOM_CSS = """
:root {
    --bg-main:    #0f1117;
    --bg-card:    #1a1d27;
    --bg-input:   #12151f;
    --border:     #2a2d3a;
    --accent:     #7c6fcd;
    --accent-dim: #3d3670;
    --text-main:  #e8e8f0;
    --text-muted: #7a7a9a;
    --text-label: #a0a0c0;
    --radius:     12px;
}
body, .gradio-container {
    background-color: var(--bg-main) !important;
    color: var(--text-main) !important;
    font-family: 'Inter', 'Segoe UI', sans-serif !important;
}
.app-header {
    text-align: center;
    padding: 36px 20px 24px;
    border-bottom: 1px solid var(--border);
    margin-bottom: 24px;
}
.app-header h1 { font-size: 26px; font-weight: 700; color: var(--text-main); margin: 0 0 6px; }
.app-header p  { color: var(--text-muted); font-size: 13px; margin: 0; }

.card {
    background: var(--bg-card);
    border: 1px solid var(--border);
    border-radius: var(--radius);
    padding: 18px 20px;
    margin-bottom: 14px;
}
.step-badge {
    display: inline-flex;
    align-items: center;
    gap: 6px;
    background: var(--accent-dim);
    color: #c4b8ff;
    font-size: 11px;
    font-weight: 600;
    letter-spacing: 0.8px;
    text-transform: uppercase;
    padding: 3px 10px;
    border-radius: 20px;
    margin-bottom: 5px;
}
.step-title { font-size: 15px; font-weight: 600; color: var(--text-main); margin: 0 0 3px; }
.step-desc  { font-size: 12px; color: var(--text-muted); margin: 0; line-height: 1.5; }

.section-label {
    font-size: 12px;
    font-weight: 700;
    letter-spacing: 1px;
    text-transform: uppercase;
    color: var(--accent);
    margin: 6px 0 10px;
}

.warn-box {
    background: #2a1f0e;
    border: 1px solid #7a4f1a;
    border-radius: 8px;
    padding: 12px 16px;
    font-size: 13px;
    color: #f5c97a;
    margin-bottom: 10px;
    line-height: 1.6;
}
.ok-box {
    background: #0e2a1a;
    border: 1px solid #1a6640;
    border-radius: 8px;
    padding: 12px 16px;
    font-size: 13px;
    color: #6ee7a8;
    margin-bottom: 10px;
    line-height: 1.6;
}
.pipeline-box {
    background: #0e1a2a;
    border: 1px solid #1a4f7a;
    border-radius: 8px;
    padding: 12px 16px;
    font-size: 13px;
    color: #7ac4f5;
    margin-bottom: 10px;
    line-height: 1.6;
}
.diag-grid {
    display: grid;
    grid-template-columns: repeat(4, 1fr);
    gap: 10px;
    margin-bottom: 10px;
}
.diag-metric {
    background: var(--bg-input);
    border: 1px solid var(--border);
    border-radius: 8px;
    padding: 10px 12px;
    text-align: center;
}
.diag-metric .val { font-size: 17px; font-weight: 700; color: var(--text-main); }
.diag-metric .lbl { font-size: 10px; color: var(--text-muted); text-transform: uppercase; letter-spacing: 0.5px; margin-top: 2px; }
.diag-pills { display: flex; flex-wrap: wrap; gap: 6px; margin-top: 4px; }
.diag-pill {
    font-size: 11px;
    background: var(--accent-dim);
    color: #c4b8ff;
    padding: 4px 10px;
    border-radius: 14px;
    font-weight: 600;
}

.gr-button-primary {
    background: var(--accent) !important;
    border: none !important;
    color: #fff !important;
    border-radius: 8px !important;
    font-weight: 600 !important;
    padding: 12px 0 !important;
    transition: opacity 0.2s !important;
}
.gr-button-primary:hover { opacity: 0.85 !important; }

.gr-textbox textarea, .gr-textbox input {
    background: var(--bg-input) !important;
    border: 1px solid var(--border) !important;
    color: var(--text-main) !important;
    border-radius: 8px !important;
    font-family: 'JetBrains Mono', 'Courier New', monospace !important;
    font-size: 12px !important;
}
label span { color: var(--text-label) !important; font-size: 12px !important; font-weight: 500 !important; }
.gr-image  { border-radius: 8px !important; border: 1px solid var(--border) !important; background: #000 !important; }
.gr-file   { background: var(--bg-input) !important; border: 1px solid var(--border) !important; border-radius: 8px !important; }
.flow-arrow { text-align: center; color: var(--accent); font-size: 18px; margin: 0; opacity: 0.6; }

.transparent-preview {
    background-color: #f0f0f0 !important;
    background-image:
        linear-gradient(45deg, #d4d4d4 25%, transparent 25%),
        linear-gradient(-45deg, #d4d4d4 25%, transparent 25%),
        linear-gradient(45deg, transparent 75%, #d4d4d4 75%),
        linear-gradient(-45deg, transparent 75%, #d4d4d4 75%) !important;
    background-size: 20px 20px !important;
    background-position: 0 0, 0 10px, 10px -10px, -10px 0px !important;
}

.final-card {
    background: linear-gradient(180deg, var(--bg-card), #161926);
    border: 1px solid var(--accent-dim);
}
"""

MATRIX_PREVIEW_SIZE = 50


# ==========================================================
# HELPER: PREVIEW MATRIX KECIL (bukan matriks penuh)
# ==========================================================
def arr_to_preview_df(arr, size=MATRIX_PREVIEW_SIZE):
    if arr is None:
        return pd.DataFrame()
    h = min(size, arr.shape[0])
    w = min(size, arr.shape[1])
    if arr.ndim == 2:
        return pd.DataFrame(arr[:h, :w])
    elif arr.ndim == 3:
        return pd.DataFrame(arr[:h, :w, 0])
    return pd.DataFrame()


# ==========================================================
# HELPER: HISTOGRAM (Terisolasi dari State Global Matplotlib)
# ==========================================================
def make_histogram_image(arr, title="Histogram"):
    fig = Figure(figsize=(5, 2.4), dpi=100)
    canvas = FigureCanvasAgg(fig)
    fig.patch.set_facecolor("#1a1d27")

    ax = fig.add_subplot(111)
    ax.set_facecolor("#12151f")

    if arr.ndim == 2:
        ax.hist(arr.ravel(), bins=64, range=(0, 255), color="#7c6fcd", alpha=0.9)
    else:
        colors = ["#ff6b6b", "#6bcb77", "#4d96ff", "#f5c97a"]
        labels_full = ["R", "G", "B", "A"]
        n_ch = arr.shape[2]
        for c in range(n_ch):
            ax.hist(arr[:, :, c].ravel(), bins=64, range=(0, 255),
                     color=colors[c % 4], alpha=0.5, label=labels_full[c % 4])
        ax.legend(fontsize=7, facecolor="#1a1d27", labelcolor="#e8e8f0")

    ax.set_title(title, color="#e8e8f0", fontsize=10)
    ax.tick_params(colors="#a0a0c0", labelsize=8)
    for spine in ax.spines.values():
        spine.set_color("#2a2d3a")
    fig.tight_layout()

    canvas.draw()
    buf = np.asarray(canvas.buffer_rgba())
    img_rgb = cv2.cvtColor(buf, cv2.COLOR_RGBA2RGB)
    return img_rgb


# ==========================================================
# HELPER: STATISTIK PIKSEL
# ==========================================================
def compute_pixel_stats(arr, active_mask=None):
    h, w = arr.shape[0], arr.shape[1]
    total_px = h * w
    lines = []
    lines.append("Resolusi        : {} × {} px".format(w, h))

    if arr.ndim == 2:
        lines.append("Channel         : 1 (Grayscale/Biner)")
        lines.append("Min Pixel       : {}".format(int(arr.min())))
        lines.append("Max Pixel       : {}".format(int(arr.max())))
        lines.append("Mean            : {:.2f}".format(float(arr.mean())))
        lines.append("Std Deviation   : {:.2f}".format(float(arr.std())))
    else:
        n_ch = arr.shape[2]
        ch_names = ["R", "G", "B", "Alpha"] if n_ch == 4 else ["R", "G", "B"]
        lines.append("Channel         : {} ({})".format(n_ch, "RGBA" if n_ch == 4 else "RGB"))
        lines.append("Min Pixel       : {}".format(int(arr.min())))
        lines.append("Max Pixel       : {}".format(int(arr.max())))
        for i, name in enumerate(ch_names):
            lines.append("Mean {:<6}   : {:.2f}".format(name, float(arr[:, :, i].mean())))
        lines.append("Std Deviation   : {:.2f} (gabungan semua channel)".format(float(arr.std())))

    if active_mask is not None:
        active_count = int(np.sum(active_mask))
        bg_count = total_px - active_count
        pct = (active_count / total_px) * 100 if total_px > 0 else 0
        lines.append("Pixel Aktif     : {:,} px".format(active_count))
        lines.append("Pixel Background: {:,} px".format(bg_count))
        lines.append("Persentase Objek: {:.2f}%".format(pct))

    return "\n".join(lines)


# ==========================================================
# ANALISIS KUALITAS GAMBAR
# ==========================================================
def analyze_image_quality(gray, h, w):
    warnings = []
    suggestions = []

    mean_bright = np.mean(gray)
    if mean_bright < 80:
        warnings.append("Gambar terlalu gelap (kecerahan {:.0f}/255)".format(mean_bright))
        suggestions.append("Foto ulang di tempat lebih terang.")
    elif mean_bright > 230:
        warnings.append("Gambar terlalu terang/overexposed ({:.0f}/255)".format(mean_bright))
        suggestions.append("Kurangi cahaya saat memotret.")

    std_dev = np.std(gray)
    if std_dev < 20:
        warnings.append("Kontras sangat rendah (std {:.1f}) — kemungkinan pensil tipis.".format(std_dev))
        suggestions.append("Gunakan pulpen lebih tebal atau perbesar kontras.")

    lap_var = cv2.Laplacian(gray, cv2.CV_64F).var()
    if lap_var < 50:
        warnings.append("Gambar buram/tidak fokus (sharpness {:.1f}).".format(lap_var))
        suggestions.append("Gunakan kamera yang lebih stabil dan fokus.")

    _, tmp = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    dark_ratio = np.sum(tmp == 0) / (h * w)
    if dark_ratio > 0.5:
        warnings.append("Lebih dari 50% area gelap — kemungkinan bayangan/background kotor.")
        suggestions.append("Foto di kertas putih bersih tanpa bayangan.")

    if w < 200 or h < 200:
        warnings.append("Resolusi terlalu kecil ({}×{} px).".format(w, h))
        suggestions.append("Upload foto minimal 400×400 px.")

    is_ok = len(warnings) == 0
    return warnings, suggestions, is_ok, mean_bright, std_dev, lap_var


def build_diagnosis_html(warnings, suggestions, is_ok, mean_bright, std_dev, lap_var, w, h, pipeline_steps):
    status_html = (
        '<div class="ok-box">✅ <strong>Gambar terdeteksi baik</strong> — siap diproses dengan pengaturan standar.</div>'
        if is_ok else
        '<div class="warn-box">⚠️ <strong>Beberapa hal perlu diperhatikan</strong> — sistem tetap menjalankan pipeline dengan penyesuaian otomatis.<ul style="margin:8px 0 0 16px;padding:0">'
        + "".join("<li style='margin-bottom:4px'>{} <span style='opacity:0.8'>{}</span></li>".format(w_, s_) for w_, s_ in zip(warnings, suggestions))
        + "</ul></div>"
    )

    metrics_html = """
    <div class="diag-grid">
      <div class="diag-metric"><div class="val">{:.0f}</div><div class="lbl">Kecerahan</div></div>
      <div class="diag-metric"><div class="val">{:.1f}</div><div class="lbl">Kontras (std)</div></div>
      <div class="diag-metric"><div class="val">{:.0f}</div><div class="lbl">Ketajaman</div></div>
      <div class="diag-metric"><div class="val">{}×{}</div><div class="lbl">Resolusi</div></div>
    </div>
    """.format(mean_bright, std_dev, lap_var, w, h)

    pills_html = ""
    if pipeline_steps:
        pills_html = '<div class="diag-pills">' + "".join(
            '<span class="diag-pill">✓ {}</span>'.format(p) for p in pipeline_steps
        ) + "</div>"

    return status_html + metrics_html + pills_html


def adaptive_preprocess(gray, std_dev, lap_var, mode, filter_method):
    processed = gray.copy()
    steps_used = []
    pipeline_flags = []

    if lap_var < 80:
        kernel_sharp = np.array([[0, -1, 0], [-1, 5, -1], [0, -1, 0]])
        processed = cv2.filter2D(processed, -1, kernel_sharp)
        steps_used.append("Sharpening (koreksi blur)")

    clahe_active = (mode == "Otomatis" and std_dev < 45)
    if clahe_active:
        clahe = cv2.createCLAHE(clipLimit=3.0, tileGridSize=(8, 8))
        processed = clahe.apply(processed)
        steps_used.append("CLAHE (perkuat kontras lokal)")
        pipeline_flags.append("CLAHE aktif")

    if mode == "Manual":
        if filter_method == "Bilateral":
            processed = cv2.bilateralFilter(processed, d=7, sigmaColor=50, sigmaSpace=50)
            steps_used.append("Bilateral Filter (manual)")
            pipeline_flags.append("Bilateral Filter (manual)")
        elif filter_method == "Gaussian":
            processed = cv2.GaussianBlur(processed, (5, 5), 0)
            steps_used.append("Gaussian Blur (manual)")
            pipeline_flags.append("Gaussian Blur (manual)")
        else:
            steps_used.append("Tanpa filter noise (manual)")
    else:
        processed = cv2.bilateralFilter(processed, d=7, sigmaColor=50, sigmaSpace=50)
        steps_used.append("Bilateral Filter (redam noise, jaga tepi)")
        pipeline_flags.append("Bilateral Filter aktif")

    return processed, steps_used, pipeline_flags


def smart_threshold(preprocessed, h0, w0, mode, threshold_method):
    pipeline_flags = []
    total_px = h0 * w0

    otsu_val, thresh_otsu = cv2.threshold(
        preprocessed, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU
    )

    if mode == "Manual":
        if threshold_method == "Otsu":
            pipeline_flags.append("Otsu Threshold (dipilih manual)")
            return thresh_otsu, "Otsu (manual)", otsu_val, pipeline_flags
        else:
            block_size = max(11, (min(h0, w0) // 20) | 1)
            thresh_adaptive = cv2.adaptiveThreshold(
                preprocessed, 255,
                cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
                cv2.THRESH_BINARY,
                block_size,
                5
            )
            pipeline_flags.append("Adaptive Threshold (dipilih manual)")
            return thresh_adaptive, "Adaptive Gaussian (manual)", otsu_val, pipeline_flags

    active_ratio_otsu = np.sum(thresh_otsu == 0) / total_px
    otsu_bad = (active_ratio_otsu < 0.002) or (active_ratio_otsu > 0.35)

    if not otsu_bad:
        pipeline_flags.append("Otsu Threshold digunakan")
        return thresh_otsu, "Otsu", otsu_val, pipeline_flags

    block_size = max(11, (min(h0, w0) // 20) | 1)
    thresh_adaptive = cv2.adaptiveThreshold(
        preprocessed, 255,
        cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
        cv2.THRESH_BINARY,
        block_size,
        5
    )
    pipeline_flags.append("Adaptive Threshold aktif (fallback dari Otsu)")
    return thresh_adaptive, "Adaptive Gaussian", otsu_val, pipeline_flags


SVG_QUALITY_EPSILON = {
    "Rendah (file kecil)": 2.0,
    "Sedang": 0.8,
    "Tinggi (detail halus)": 0.3,
}

# ==========================================================
# PIPELINE UTAMA
# ==========================================================
def process_signature(image, mode, threshold_method, filter_method, svg_quality):
    NOUT = 41
    if image is None:
        return (None,) * NOUT

    tmp_dir = tempfile.mkdtemp()
    epsilon = SVG_QUALITY_EPSILON.get(svg_quality, 0.8)

    img = cv2.cvtColor(image, cv2.COLOR_RGB2BGR)
    h0, w0 = img.shape[:2]

    # ── ANALISIS KUALITAS ─────────────────────────────────
    gray_raw = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    std_dev  = np.std(gray_raw)
    lap_var  = cv2.Laplacian(gray_raw, cv2.CV_64F).var()

    warnings, suggestions, is_ok, mean_bright, std_dev2, lap_var2 = analyze_image_quality(gray_raw, h0, w0)

    # ── TAHAP 1: GRAYSCALE ────────────────────────────────
    gray = gray_raw.copy()
    gray_display = cv2.cvtColor(gray, cv2.COLOR_GRAY2RGB)
    hist_gray = make_histogram_image(gray, "Distribusi Pixel — Grayscale")
    stats_gray = compute_pixel_stats(gray)
    df_gray = arr_to_preview_df(gray)
    info_gray = "Resolusi : {} × {} px\nChannel  : 1 (Grayscale)\nRata-rata: {:.0f}\nStd dev  : {:.1f}".format(w0, h0, np.mean(gray), std_dev)

    # ── TAHAP 2: PRE-PROCESSING + THRESHOLD ───────────────
    preprocessed, steps_used, pre_flags = adaptive_preprocess(gray, std_dev, lap_var, mode, filter_method)
    thresh, method_name, otsu_val, thresh_flags = smart_threshold(preprocessed, h0, w0, mode, threshold_method)

    pipeline_flags = pre_flags + thresh_flags
    diagnosis_html = build_diagnosis_html(warnings, suggestions, is_ok, mean_bright, std_dev, lap_var, w0, h0, pipeline_flags)

    thresh_display = cv2.cvtColor(thresh, cv2.COLOR_GRAY2RGB)
    hist_thresh = make_histogram_image(thresh, "Distribusi Pixel — Threshold")
    thresh_active_mask = (thresh == 255)
    stats_thresh = compute_pixel_stats(thresh, active_mask=thresh_active_mask)
    df_thresh = arr_to_preview_df(thresh)
    adaptasi_info = "\n".join(["• " + s for s in steps_used]) if steps_used else "• Tidak diperlukan"
    info_thresh = "Mode        : {}\nMetode      : {}\nOtsu (ref)  : {:.0f}\nPre-process :\n{}".format(
        mode, method_name, otsu_val, adaptasi_info
    )

    # ── TAHAP 3: MORFOLOGI ────────────────────────────────
    inv = cv2.bitwise_not(thresh)
    k_open = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (3, 3))
    opened = cv2.morphologyEx(inv, cv2.MORPH_OPEN, k_open, iterations=1)
    closed = opened.copy()

    morph_display = cv2.cvtColor(cv2.bitwise_not(closed), cv2.COLOR_GRAY2RGB)
    hist_morph = make_histogram_image(closed, "Distribusi Pixel — Morfologi")
    morph_active_mask = (closed == 255)
    stats_morph = compute_pixel_stats(closed, active_mask=morph_active_mask)
    df_morph = arr_to_preview_df(closed)
    noise_before = np.sum(inv == 255)
    noise_after = np.sum(closed == 255)
    info_morph = "Opening  : hapus noise kecil\nClosing  : nonaktif (jaga stroke tipis)\nKernel   : Ellipse 3×3\nSebelum  : {:,} px\nSesudah  : {:,} px\nDihapus  : {:,} px".format(
        noise_before, noise_after, max(0, noise_before - noise_after))

    # ── TAHAP 4: DETEKSI KONTUR ────────────────────────────
    contours, hierarchy = cv2.findContours(closed, cv2.RETR_TREE, cv2.CHAIN_APPROX_TC89_KCOS)
    total_px = h0 * w0
    min_area = max(6, total_px * 0.00005)
    valid_contours = [c for c in contours if cv2.contourArea(c) > min_area]

    contour_vis = img.copy()
    cv2.drawContours(contour_vis, valid_contours, -1, (205, 111, 124), 2)
    contour_display = cv2.cvtColor(contour_vis, cv2.COLOR_BGR2RGB)
    hist_contour = make_histogram_image(contour_display, "Distribusi Pixel — Overlay Kontur")
    stats_contour = compute_pixel_stats(contour_display)
    df_contour = arr_to_preview_df(contour_display)
    info_contour = "Total kontur : {}\nKontur valid : {} (area > {:.1f}px)\nNoise dibuang: {}\nMetode approx: TC89_KCOS".format(
        len(contours), len(valid_contours), min_area, len(contours) - len(valid_contours))

    # ── TAHAP 5: CROP & RESIZE 400×400 ────────────────────
    coords = cv2.findNonZero(closed)
    if coords is None:
        empty_img = np.ones((400, 400, 3), dtype=np.uint8) * 30
        empty_hist = make_histogram_image(np.zeros((10, 10), dtype=np.uint8), "Tidak ada data")
        empty_df = pd.DataFrame()
        err = "Tidak ada objek terdeteksi.\nCoba upload ulang."
        return (
            diagnosis_html,
            gray_display, hist_gray, stats_gray, info_gray, df_gray,
            thresh_display, hist_thresh, stats_thresh, info_thresh, df_thresh,
            morph_display, hist_morph, stats_morph, info_morph, df_morph,
            contour_display, hist_contour, stats_contour, info_contour, df_contour,
            empty_img, empty_hist, err, err, empty_df,
            empty_img, empty_hist, err, err, empty_df,
            empty_img, empty_img, None, err,
            None, err,
            None, None, None, None,
        )

    x, y, w, h = cv2.boundingRect(coords)
    pad = 20
    x1, y1 = max(0, x - pad), max(0, y - pad)
    x2, y2 = min(closed.shape[1], x + w + pad), min(closed.shape[0], y + h + pad)
    cropped = closed[y1:y2, x1:x2]

    TARGET = 400
    scale = min(TARGET / cropped.shape[1], TARGET / cropped.shape[0])
    nw, nh = int(cropped.shape[1] * scale), int(cropped.shape[0] * scale)
    resized = cv2.resize(cropped, (nw, nh), interpolation=cv2.INTER_AREA)
    _, resized = cv2.threshold(resized, 127, 255, cv2.THRESH_BINARY)

    canvas = np.zeros((TARGET, TARGET), dtype=np.uint8)
    ox, oy = (TARGET - nw) // 2, (TARGET - nh) // 2
    canvas[oy:oy+nh, ox:ox+nw] = resized

    final_png = cv2.bitwise_not(canvas)
    final_display = cv2.cvtColor(final_png, cv2.COLOR_GRAY2RGB)
    hist_final = make_histogram_image(final_png, "Distribusi Pixel — Crop & Resize")
    final_active_mask = (canvas == 255)
    stats_final = compute_pixel_stats(final_png, active_mask=final_active_mask)
    df_final = arr_to_preview_df(final_png)
    info_resize = "Bounding box : {}×{} px\nPadding      : {} px\nSkala resize : {:.3f}×\nKanvas akhir : {}×{} px".format(
        w, h, pad, scale, TARGET, TARGET)

    # ── TAHAP 5.5: BACKGROUND REMOVAL ──────────────────────
    # Konversi RGBA yang aman secara struktur untuk UI Web
    rgba = cv2.cvtColor(final_png, cv2.COLOR_GRAY2RGBA)
    rgba[final_png == 255] = [255, 255, 255, 0]
    rgba[final_png == 0] = [0, 0, 0, 255]

    hist_transparent = make_histogram_image(rgba, "Distribusi Pixel — RGBA")
    transparent_active_mask = (canvas == 255)
    stats_transparent = compute_pixel_stats(rgba, active_mask=transparent_active_mask)
    df_transparent = arr_to_preview_df(rgba)

    checker = np.zeros((TARGET, TARGET, 3), dtype=np.uint8)
    cell = 20
    for cy in range(0, TARGET, cell):
        for cx in range(0, TARGET, cell):
            color = 240 if ((cx // cell) + (cy // cell)) % 2 == 0 else 215
            checker[cy:cy+cell, cx:cx+cell] = (color, color, color)

    alpha = rgba[:, :, 3:4].astype(np.float32) / 255.0
    rgb_fg = rgba[:, :, :3].astype(np.float32)
    transparent_preview = (rgb_fg * alpha + checker.astype(np.float32) * (1 - alpha)).astype(np.uint8)

    transparent_pixel_count = int(np.sum(final_png == 0))
    info_transparent = "Area objek    : {:,} px\nArea transparan: {:,} px\nDimensi       : {}×{} px (RGBA)".format(
        transparent_pixel_count, TARGET*TARGET - transparent_pixel_count, TARGET, TARGET
    )

    # ── TAHAP 6: EXPORT SVG ─────────────────────────────────
    path_svg = os.path.join(tmp_dir, "signature_vectorized.svg")
    dwg = svgwrite.Drawing(path_svg, size=("400px", "400px"), profile="full")

    svg_contours, svg_hier = cv2.findContours(canvas, cv2.RETR_TREE, cv2.CHAIN_APPROX_TC89_KCOS)
    path_count = 0
    if svg_hier is not None:
        for cnt in svg_contours:
            if cv2.contourArea(cnt) > min_area:
                approx = cv2.approxPolyDP(cnt, epsilon=epsilon, closed=True)
                pts = [(int(p[0][0]), int(p[0][1])) for p in approx]
                if len(pts) >= 3:
                    dwg.add(dwg.polygon(points=pts, fill="black", stroke="black", stroke_width=0.3))
                    path_count += 1
    dwg.save()
    info_svg = "Format      : SVG\nKualitas    : {} (epsilon={})\nPath/polygon: {}\nViewbox     : 400×400".format(
        svg_quality, epsilon, path_count)

    # ── SIMPAN FILE ──────────────────────────────────────────
    path_clean = os.path.join(tmp_dir, "signature_clean.png")
    cv2.imwrite(path_clean, final_png)

    path_transparent = os.path.join(tmp_dir, "signature_transparent.png")
    # OpenCV imwrite membutuhkan format BGRA secara native untuk disave ke disk
    cv2.imwrite(path_transparent, cv2.cvtColor(rgba, cv2.COLOR_RGBA2BGRA))

    path_csv = os.path.join(tmp_dir, "signature_matrix.csv")
    pd.DataFrame(canvas).to_csv(path_csv, index=False)

    path_zip = os.path.join(tmp_dir, "hasil_signature.zip")
    with zipfile.ZipFile(path_zip, "w") as zf:
        zf.write(path_clean, arcname="signature_clean.png")
        zf.write(path_transparent, arcname="signature_transparent.png")
        zf.write(path_svg, arcname="signature_vectorized.svg")
        zf.write(path_csv, arcname="signature_matrix.csv")

    # ── OUTPUT FINAL (ringkasan akhir) ─────────────────────
    final_summary_stats = (
        "Jumlah kontur valid : {}\n"
        "Ukuran hasil        : {}×{} px\n"
        "Jumlah path SVG     : {}\n"
        "Kualitas SVG        : {}"
    ).format(len(valid_contours), TARGET, TARGET, path_count, svg_quality)

    return (
        diagnosis_html,
        gray_display,    hist_gray,        stats_gray,        info_gray,    df_gray,
        thresh_display,  hist_thresh,      stats_thresh,      info_thresh,  df_thresh,
        morph_display,   hist_morph,       stats_morph,       info_morph,   df_morph,
        contour_display, hist_contour,     stats_contour,     info_contour, df_contour,
        final_display,   hist_final,       stats_final,       info_resize,  df_final,
        transparent_preview, hist_transparent, stats_transparent, info_transparent, df_transparent,
        final_display, transparent_preview, path_svg, final_summary_stats,
        path_svg, info_svg,
        path_clean, path_transparent, path_csv, path_zip,
    )


# ==========================================================
# UI
# ==========================================================
with gr.Blocks(css=CUSTOM_CSS, title="Signature to Vector") as demo:

    gr.HTML("""
    <div class="app-header">
        <h1>✍️ Mini Lab Pengolahan Citra Tanda Tangan</h1>
        <p>Upload → Diagnosis → Grayscale → Threshold → Morfologi → Kontur → Crop & Resize → Background Removal → SVG</p>
    </div>
    """)

    with gr.Row(equal_height=False):

        # ════════════════════════════════════════════════
        # SIDEBAR KIRI
        # ════════════════════════════════════════════════
        with gr.Column(scale=1, min_width=280):

            # Card 1: Upload & Proses
            gr.HTML('<div class="card">')
            gr.HTML('<div class="step-badge">📥 Input</div>')
            gr.HTML('<div class="step-title">Upload Tanda Tangan</div>')
            gr.HTML('<div class="step-desc">Foto atau scan tanda tangan — pulpen maupun pensil.</div>')
            input_image = gr.Image(type="numpy", label="Citra Tanda Tangan")
            img_info_box = gr.Textbox(label="Info Gambar", lines=2, interactive=False, placeholder="Ukuran gambar akan muncul di sini...")
            btn = gr.Button("▶  Jalankan Pipeline", variant="primary")
            gr.HTML('</div>')

            # Card 2: Pengaturan Proses
            gr.HTML('<div class="card">')
            gr.HTML('<div class="step-badge">⚙️ Pengaturan</div>')
            gr.HTML('<div class="step-title">Pengaturan Proses</div>')
            gr.HTML('<div class="step-desc">Mode Otomatis cocok untuk kebanyakan kasus. Mode Manual memberi kontrol penuh.</div>')

            mode_radio = gr.Radio(
                choices=["Otomatis", "Manual"],
                value="Otomatis",
                label="Mode Proses"
            )

            with gr.Accordion("Pengaturan Lanjutan", open=False):
                threshold_method = gr.Dropdown(
                    choices=["Otsu", "Adaptive Gaussian"],
                    value="Otsu",
                    label="Threshold Method",
                    info="Hanya berlaku jika Mode = Manual"
                )
                filter_method = gr.Dropdown(
                    choices=["Bilateral", "Gaussian", "Tidak ada"],
                    value="Bilateral",
                    label="Filter Method",
                    info="Hanya berlaku jika Mode = Manual"
                )
                svg_quality = gr.Dropdown(
                    choices=["Rendah (file kecil)", "Sedang", "Tinggi (detail halus)"],
                    value="Sedang",
                    label="SVG Quality"
                )
            gr.HTML('</div>')

            # Card 3: Download
            gr.HTML('<div class="card">')
            gr.HTML('<div class="step-badge">📦 Unduh Hasil</div>')
            out_zip  = gr.File(label="📁 Paket Lengkap (.zip)")
            out_png  = gr.File(label="🖼 PNG Bersih")
            out_trns = gr.File(label="🫧 PNG Transparan")
            out_svg_dl = gr.File(label="✒️ SVG Vektor")
            out_csv  = gr.File(label="📊 Matriks (.csv)")
            gr.HTML('</div>')

        # ════════════════════════════════════════════════
        # AREA KANAN
        # ════════════════════════════════════════════════
        with gr.Column(scale=3):

            # ── 1. DIAGNOSIS (paling atas) ──
            gr.HTML('<div class="section-label">1 · Diagnosis Kualitas Gambar</div>')
            diagnosis_out = gr.HTML(value="<div class='warn-box' style='opacity:0.4'>Hasil diagnosis akan muncul di sini setelah gambar diproses...</div>")

            gr.HTML('<div class="flow-arrow">↓</div>')

            # ── 2. TAHAP PROSES ──
            gr.HTML('<div class="section-label">2 · Tahap Proses</div>')

            def build_stage_block(badge, title, desc, img_label, is_transparent=False):
                gr.HTML(f"""
                <div class="card">
                  <div class="step-badge">{badge}</div>
                  <div class="step-title">{title}</div>
                  <div class="step-desc">{desc}</div>
                </div>
                """)
                with gr.Tabs():
                    with gr.TabItem("🖼 Hasil Gambar"):
                        img_comp = gr.Image(
                            label=img_label,
                            type="numpy",
                            height=300,
                            elem_classes=["transparent-preview"] if is_transparent else None
                        )
                    with gr.TabItem("📈 Histogram & Statistik"):
                        hist_comp = gr.Image(label="Histogram Distribusi Pixel", type="numpy", height=220)
                        stats_comp = gr.Textbox(label="Statistik Pixel", lines=8, interactive=False)
                with gr.Accordion("🔧 Detail Teknis", open=False):
                    info_comp = gr.Textbox(label="Informasi Proses", lines=6, interactive=False)
                    df_comp = gr.DataFrame(
                        label=f"Preview Matrix {MATRIX_PREVIEW_SIZE}×{MATRIX_PREVIEW_SIZE} (pojok kiri-atas — unduh CSV untuk data lengkap)",
                        interactive=False
                    )
                gr.HTML('<div class="flow-arrow">↓</div>')
                return img_comp, hist_comp, stats_comp, info_comp, df_comp

            (img_gray, hist_gray_comp, stats_gray_comp, info_gray, df_gray_comp) = build_stage_block(
                "Tahap 1", "Grayscale", "Citra diubah ke 1-channel (0–255) agar threshold lebih akurat.", "Hasil Grayscale"
            )
            (img_thresh, hist_thresh_comp, stats_thresh_comp, info_thresh, df_thresh_comp) = build_stage_block(
                "Tahap 2", "Thresholding", "Binarisasi citra — otomatis memilih metode terbaik atau sesuai pilihan manual.", "Hasil Binarisasi"
            )
            (img_morph, hist_morph_comp, stats_morph_comp, info_morph, df_morph_comp) = build_stage_block(
                "Tahap 3", "Morphology", "Membersihkan noise kecil sambil menjaga goresan tipis tanda tangan.", "Setelah Morfologi"
            )
            (img_contour, hist_contour_comp, stats_contour_comp, info_contour, df_contour_comp) = build_stage_block(
                "Tahap 4", "Contour Detection", "Melacak batas setiap goresan dengan ambang area adaptif.", "Overlay Kontur"
            )
            (img_final, hist_final_comp, stats_final_comp, info_final, df_final_comp) = build_stage_block(
                "Tahap 5", "Crop & Resize", "Memotong area aktif dan menormalkan ke kanvas 400×400 px.", "Hasil 400×400"
            )
            (img_transparent, hist_transparent_comp, stats_transparent_comp, info_transparent, df_transparent_comp) = build_stage_block(
                "Tahap 6", "Background Removal", "Menghapus latar belakang — area putih menjadi transparan (alpha=0).", "Preview PNG Transparan", is_transparent=True
            )

            # ── 3. OUTPUT FINAL ──
            gr.HTML('<div class="section-label">3 · Output Final</div>')
            gr.HTML("""
            <div class="card final-card">
              <div class="step-badge">✅ Selesai</div>
              <div class="step-title">Hasil Akhir Tanda Tangan Digital</div>
              <div class="step-desc">Pratinjau hasil akhir dalam tiga format. Unduh melalui panel kiri.</div>
            </div>
            """)
            with gr.Row():
                final_png_preview = gr.Image(label="PNG Final", type="numpy", height=260)
                final_transparent_preview = gr.Image(label="PNG Transparan", type="numpy", height=260, elem_classes=["transparent-preview"])
                final_svg_preview = gr.File(label="File SVG Vektor")
            final_stats_box = gr.Textbox(label="Ringkasan Hasil", lines=4, interactive=False)

            with gr.Accordion("🔧 Detail Teknis SVG Export", open=False):
                info_svg = gr.Textbox(label="Informasi SVG", lines=6, interactive=False)

    # ── Update info ukuran gambar saat upload ──
    def show_image_info(image):
        if image is None:
            return ""
        h, w = image.shape[:2]
        return f"Resolusi: {w} × {h} px\nChannel : {image.shape[2] if image.ndim == 3 else 1}"

    input_image.change(fn=show_image_info, inputs=[input_image], outputs=[img_info_box])

    # 41 Total Inputs and Outputs matched
    btn.click(
        fn=process_signature,
        inputs=[input_image, mode_radio, threshold_method, filter_method, svg_quality],
        outputs=[
            diagnosis_out,
            img_gray, hist_gray_comp, stats_gray_comp, info_gray, df_gray_comp,
            img_thresh, hist_thresh_comp, stats_thresh_comp, info_thresh, df_thresh_comp,
            img_morph, hist_morph_comp, stats_morph_comp, info_morph, df_morph_comp,
            img_contour, hist_contour_comp, stats_contour_comp, info_contour, df_contour_comp,
            img_final, hist_final_comp, stats_final_comp, info_final, df_final_comp,
            img_transparent, hist_transparent_comp, stats_transparent_comp, info_transparent, df_transparent_comp,
            final_png_preview, final_transparent_preview, final_svg_preview, final_stats_box,
            out_svg_dl, info_svg,
            out_png, out_trns, out_csv, out_zip
        ]
    )

demo.launch(share=True)
```

## Helper Functions

```python
# ==========================================================
# HELPER: PREVIEW MATRIX KECIL (bukan matriks penuh)
# ==========================================================
def arr_to_preview_df(arr, size=MATRIX_PREVIEW_SIZE):
    if arr is None:
        return pd.DataFrame()
    h = min(size, arr.shape[0])
    w = min(size, arr.shape[1])
    if arr.ndim == 2:
        return pd.DataFrame(arr[:h, :w])
    elif arr.ndim == 3:
        return pd.DataFrame(arr[:h, :w, 0])
    return pd.DataFrame()


# ==========================================================
# HELPER: HISTOGRAM (Terisolasi dari State Global Matplotlib)
# ==========================================================
def make_histogram_image(arr, title="Histogram"):
    fig = Figure(figsize=(5, 2.4), dpi=100)
    canvas = FigureCanvasAgg(fig)
    fig.patch.set_facecolor("#1a1d27")

    ax = fig.add_subplot(111)
    ax.set_facecolor("#12151f")

    if arr.ndim == 2:
        ax.hist(arr.ravel(), bins=64, range=(0, 255), color="#7c6fcd", alpha=0.9)
    else:
        colors = ["#ff6b6b", "#6bcb77", "#4d96ff", "#f5c97a"]
        labels_full = ["R", "G", "B", "A"]
        n_ch = arr.shape[2]
        for c in range(n_ch):
            ax.hist(arr[:, :, c].ravel(), bins=64, range=(0, 255),
                     color=colors[c % 4], alpha=0.5, label=labels_full[c % 4])
        ax.legend(fontsize=7, facecolor="#1a1d27", labelcolor="#e8e8f0")

    ax.set_title(title, color="#e8e8f0", fontsize=10)
    ax.tick_params(colors="#a0a0c0", labelsize=8)
    for spine in ax.spines.values():
        spine.set_color("#2a2d3a")
    fig.tight_layout()

    canvas.draw()
    buf = np.asarray(canvas.buffer_rgba())
    img_rgb = cv2.cvtColor(buf, cv2.COLOR_RGBA2RGB)
    return img_rgb


# ==========================================================
# HELPER: STATISTIK PIKSEL
# ==========================================================
def compute_pixel_stats(arr, active_mask=None):
    h, w = arr.shape[0], arr.shape[1]
    total_px = h * w
    lines = []
    lines.append("Resolusi        : {} × {} px".format(w, h))

    if arr.ndim == 2:
        lines.append("Channel         : 1 (Grayscale/Biner)")
        lines.append("Min Pixel       : {}".format(int(arr.min())))
        lines.append("Max Pixel       : {}".format(int(arr.max())))
        lines.append("Mean            : {:.2f}".format(float(arr.mean())))
        lines.append("Std Deviation   : {:.2f}".format(float(arr.std())))
    else:
        n_ch = arr.shape[2]
        ch_names = ["R", "G", "B", "Alpha"] if n_ch == 4 else ["R", "G", "B"]
        lines.append("Channel         : {} ({})".format(n_ch, "RGBA" if n_ch == 4 else "RGB"))
        lines.append("Min Pixel       : {}".format(int(arr.min())))
        lines.append("Max Pixel       : {}".format(int(arr.max())))
        for i, name in enumerate(ch_names):
            lines.append("Mean {:<6}   : {:.2f}".format(name, float(arr[:, :, i].mean())))
        lines.append("Std Deviation   : {:.2f} (gabungan semua channel)".format(float(arr.std())))

    if active_mask is not None:
        active_count = int(np.sum(active_mask))
        bg_count = total_px - active_count
        pct = (active_count / total_px) * 100 if total_px > 0 else 0
        lines.append("Pixel Aktif     : {:,} px".format(active_count))
        lines.append("Pixel Background: {:,} px".format(bg_count))
        lines.append("Persentase Objek: {:.2f}%".format(pct))

    return "\n".join(lines)


# ==========================================================
# ANALISIS KUALITAS GAMBAR
# ==========================================================
def analyze_image_quality(gray, h, w):
    warnings = []
    suggestions = []

    mean_bright = np.mean(gray)
    if mean_bright < 80:
        warnings.append("Gambar terlalu gelap (kecerahan {:.0f}/255)".format(mean_bright))
        suggestions.append("Foto ulang di tempat lebih terang.")
    elif mean_bright > 230:
        warnings.append("Gambar terlalu terang/overexposed ({:.0f}/255)".format(mean_bright))
        suggestions.append("Kurangi cahaya saat memotret.")

    std_dev = np.std(gray)
    if std_dev < 20:
        warnings.append("Kontras sangat rendah (std {:.1f}) — kemungkinan pensil tipis.".format(std_dev))
        suggestions.append("Gunakan pulpen lebih tebal atau perbesar kontras.")

    lap_var = cv2.Laplacian(gray, cv2.CV_64F).var()
    if lap_var < 50:
        warnings.append("Gambar buram/tidak fokus (sharpness {:.1f}).".format(lap_var))
        suggestions.append("Gunakan kamera yang lebih stabil dan fokus.")

    _, tmp = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    dark_ratio = np.sum(tmp == 0) / (h * w)
    if dark_ratio > 0.5:
        warnings.append("Lebih dari 50% area gelap — kemungkinan bayangan/background kotor.")
        suggestions.append("Foto di kertas putih bersih tanpa bayangan.")

    if w < 200 or h < 200:
        warnings.append("Resolusi terlalu kecil ({}×{} px).".format(w, h))
        suggestions.append("Upload foto minimal 400×400 px.")

    is_ok = len(warnings) == 0
    return warnings, suggestions, is_ok, mean_bright, std_dev, lap_var


def build_diagnosis_html(warnings, suggestions, is_ok, mean_bright, std_dev, lap_var, w, h, pipeline_steps):
    status_html = (
        '<div class="ok-box">✅ <strong>Gambar terdeteksi baik</strong> — siap diproses dengan pengaturan standar.</div>'
        if is_ok else
        '<div class="warn-box">⚠️ <strong>Beberapa hal perlu diperhatikan</strong> — sistem tetap menjalankan pipeline dengan penyesuaian otomatis.<ul style="margin:8px 0 0 16px;padding:0">'
        + "".join("<li style='margin-bottom:4px'>{} <span style='opacity:0.8'>{}</span></li>".format(w_, s_) for w_, s_ in zip(warnings, suggestions))
        + "</ul></div>"
    )

    metrics_html = """
    <div class="diag-grid">
      <div class="diag-metric"><div class="val">{:.0f}</div><div class="lbl">Kecerahan</div></div>
      <div class="diag-metric"><div class="val">{:.1f}</div><div class="lbl">Kontras (std)</div></div>
      <div class="diag-metric"><div class="val">{:.0f}</div><div class="lbl">Ketajaman</div></div>
      <div class="diag-metric"><div class="val">{}×{}</div><div class="lbl">Resolusi</div></div>
    </div>
    """.format(mean_bright, std_dev, lap_var, w, h)

    pills_html = ""
    if pipeline_steps:
        pills_html = '<div class="diag-pills">' + "".join(
            '<span class="diag-pill">✓ {}</span>'.format(p) for p in pipeline_steps
        ) + "</div>"

    return status_html + metrics_html + pills_html


def adaptive_preprocess(gray, std_dev, lap_var, mode, filter_method):
    processed = gray.copy()
    steps_used = []
    pipeline_flags = []

    if lap_var < 80:
        kernel_sharp = np.array([[0, -1, 0], [-1, 5, -1], [0, -1, 0]])
        processed = cv2.filter2D(processed, -1, kernel_sharp)
        steps_used.append("Sharpening (koreksi blur)")

    clahe_active = (mode == "Otomatis" and std_dev < 45)
    if clahe_active:
        clahe = cv2.createCLAHE(clipLimit=3.0, tileGridSize=(8, 8))
        processed = clahe.apply(processed)
        steps_used.append("CLAHE (perkuat kontras lokal)")
        pipeline_flags.append("CLAHE aktif")

    if mode == "Manual":
        if filter_method == "Bilateral":
            processed = cv2.bilateralFilter(processed, d=7, sigmaColor=50, sigmaSpace=50)
            steps_used.append("Bilateral Filter (manual)")
            pipeline_flags.append("Bilateral Filter (manual)")
        elif filter_method == "Gaussian":
            processed = cv2.GaussianBlur(processed, (5, 5), 0)
            steps_used.append("Gaussian Blur (manual)")
            pipeline_flags.append("Gaussian Blur (manual)")
        else:
            steps_used.append("Tanpa filter noise (manual)")
    else:
        processed = cv2.bilateralFilter(processed, d=7, sigmaColor=50, sigmaSpace=50)
        steps_used.append("Bilateral Filter (redam noise, jaga tepi)")
        pipeline_flags.append("Bilateral Filter aktif")

    return processed, steps_used, pipeline_flags


def smart_threshold(preprocessed, h0, w0, mode, threshold_method):
    pipeline_flags = []
    total_px = h0 * w0

    otsu_val, thresh_otsu = cv2.threshold(
        preprocessed, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU
    )

    if mode == "Manual":
        if threshold_method == "Otsu":
            pipeline_flags.append("Otsu Threshold (dipilih manual)")
            return thresh_otsu, "Otsu (manual)", otsu_val, pipeline_flags
        else:
            block_size = max(11, (min(h0, w0) // 20) | 1)
            thresh_adaptive = cv2.adaptiveThreshold(
                preprocessed, 255,
                cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
                cv2.THRESH_BINARY,
                block_size,
                5
            )
            pipeline_flags.append("Adaptive Threshold (dipilih manual)")
            return thresh_adaptive, "Adaptive Gaussian (manual)", otsu_val, pipeline_flags

    active_ratio_otsu = np.sum(thresh_otsu == 0) / total_px
    otsu_bad = (active_ratio_otsu < 0.002) or (active_ratio_otsu > 0.35)

    if not otsu_bad:
        pipeline_flags.append("Otsu Threshold digunakan")
        return thresh_otsu, "Otsu", otsu_val, pipeline_flags

    block_size = max(11, (min(h0, w0) // 20) | 1)
    thresh_adaptive = cv2.adaptiveThreshold(
        preprocessed, 255,
        cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
        cv2.THRESH_BINARY,
        block_size,
        5
    )
    pipeline_flags.append("Adaptive Threshold aktif (fallback dari Otsu)")
    return thresh_adaptive, "Adaptive Gaussian", otsu_val, pipeline_flags
```

## Pipeline utama

```python
SVG_QUALITY_EPSILON = {
    "Rendah (file kecil)": 2.0,
    "Sedang": 0.8,
    "Tinggi (detail halus)": 0.3,
}

# ==========================================================
# PIPELINE UTAMA
# ==========================================================
def process_signature(image, mode, threshold_method, filter_method, svg_quality):
    NOUT = 41
    if image is None:
        return (None,) * NOUT

    tmp_dir = tempfile.mkdtemp()
    epsilon = SVG_QUALITY_EPSILON.get(svg_quality, 0.8)

    img = cv2.cvtColor(image, cv2.COLOR_RGB2BGR)
    h0, w0 = img.shape[:2]

    # ── ANALISIS KUALITAS ─────────────────────────────────
    gray_raw = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    std_dev  = np.std(gray_raw)
    lap_var  = cv2.Laplacian(gray_raw, cv2.CV_64F).var()

    warnings, suggestions, is_ok, mean_bright, std_dev2, lap_var2 = analyze_image_quality(gray_raw, h0, w0)

    # ── TAHAP 1: GRAYSCALE ────────────────────────────────
    gray = gray_raw.copy()
    gray_display = cv2.cvtColor(gray, cv2.COLOR_GRAY2RGB)
    hist_gray = make_histogram_image(gray, "Distribusi Pixel — Grayscale")
    stats_gray = compute_pixel_stats(gray)
    df_gray = arr_to_preview_df(gray)
    info_gray = "Resolusi : {} × {} px\nChannel  : 1 (Grayscale)\nRata-rata: {:.0f}\nStd dev  : {:.1f}".format(w0, h0, np.mean(gray), std_dev)

    # ── TAHAP 2: PRE-PROCESSING + THRESHOLD ───────────────
    preprocessed, steps_used, pre_flags = adaptive_preprocess(gray, std_dev, lap_var, mode, filter_method)
    thresh, method_name, otsu_val, thresh_flags = smart_threshold(preprocessed, h0, w0, mode, threshold_method)

    pipeline_flags = pre_flags + thresh_flags
    diagnosis_html = build_diagnosis_html(warnings, suggestions, is_ok, mean_bright, std_dev, lap_var, w0, h0, pipeline_flags)

    thresh_display = cv2.cvtColor(thresh, cv2.COLOR_GRAY2RGB)
    hist_thresh = make_histogram_image(thresh, "Distribusi Pixel — Threshold")
    thresh_active_mask = (thresh == 255)
    stats_thresh = compute_pixel_stats(thresh, active_mask=thresh_active_mask)
    df_thresh = arr_to_preview_df(thresh)
    adaptasi_info = "\n".join(["• " + s for s in steps_used]) if steps_used else "• Tidak diperlukan"
    info_thresh = "Mode        : {}\nMetode      : {}\nOtsu (ref)  : {:.0f}\nPre-process :\n{}".format(
        mode, method_name, otsu_val, adaptasi_info
    )

    # ── TAHAP 3: MORFOLOGI ────────────────────────────────
    inv = cv2.bitwise_not(thresh)
    k_open = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (3, 3))
    opened = cv2.morphologyEx(inv, cv2.MORPH_OPEN, k_open, iterations=1)
    closed = opened.copy()

    morph_display = cv2.cvtColor(cv2.bitwise_not(closed), cv2.COLOR_GRAY2RGB)
    hist_morph = make_histogram_image(closed, "Distribusi Pixel — Morfologi")
    morph_active_mask = (closed == 255)
    stats_morph = compute_pixel_stats(closed, active_mask=morph_active_mask)
    df_morph = arr_to_preview_df(closed)
    noise_before = np.sum(inv == 255)
    noise_after = np.sum(closed == 255)
    info_morph = "Opening  : hapus noise kecil\nClosing  : nonaktif (jaga stroke tipis)\nKernel   : Ellipse 3×3\nSebelum  : {:,} px\nSesudah  : {:,} px\nDihapus  : {:,} px".format(
        noise_before, noise_after, max(0, noise_before - noise_after))

    # ── TAHAP 4: DETEKSI KONTUR ────────────────────────────
    contours, hierarchy = cv2.findContours(closed, cv2.RETR_TREE, cv2.CHAIN_APPROX_TC89_KCOS)
    total_px = h0 * w0
    min_area = max(6, total_px * 0.00005)
    valid_contours = [c for c in contours if cv2.contourArea(c) > min_area]

    contour_vis = img.copy()
    cv2.drawContours(contour_vis, valid_contours, -1, (205, 111, 124), 2)
    contour_display = cv2.cvtColor(contour_vis, cv2.COLOR_BGR2RGB)
    hist_contour = make_histogram_image(contour_display, "Distribusi Pixel — Overlay Kontur")
    stats_contour = compute_pixel_stats(contour_display)
    df_contour = arr_to_preview_df(contour_display)
    info_contour = "Total kontur : {}\nKontur valid : {} (area > {:.1f}px)\nNoise dibuang: {}\nMetode approx: TC89_KCOS".format(
        len(contours), len(valid_contours), min_area, len(contours) - len(valid_contours))

    # ── TAHAP 5: CROP & RESIZE 400×400 ────────────────────
    coords = cv2.findNonZero(closed)
    if coords is None:
        empty_img = np.ones((400, 400, 3), dtype=np.uint8) * 30
        empty_hist = make_histogram_image(np.zeros((10, 10), dtype=np.uint8), "Tidak ada data")
        empty_df = pd.DataFrame()
        err = "Tidak ada objek terdeteksi.\nCoba upload ulang."
        return (
            diagnosis_html,
            gray_display, hist_gray, stats_gray, info_gray, df_gray,
            thresh_display, hist_thresh, stats_thresh, info_thresh, df_thresh,
            morph_display, hist_morph, stats_morph, info_morph, df_morph,
            contour_display, hist_contour, stats_contour, info_contour, df_contour,
            empty_img, empty_hist, err, err, empty_df,
            empty_img, empty_hist, err, err, empty_df,
            empty_img, empty_img, None, err,
            None, err,
            None, None, None, None,
        )

    x, y, w, h = cv2.boundingRect(coords)
    pad = 20
    x1, y1 = max(0, x - pad), max(0, y - pad)
    x2, y2 = min(closed.shape[1], x + w + pad), min(closed.shape[0], y + h + pad)
    cropped = closed[y1:y2, x1:x2]

    TARGET = 400
    scale = min(TARGET / cropped.shape[1], TARGET / cropped.shape[0])
    nw, nh = int(cropped.shape[1] * scale), int(cropped.shape[0] * scale)
    resized = cv2.resize(cropped, (nw, nh), interpolation=cv2.INTER_AREA)
    _, resized = cv2.threshold(resized, 127, 255, cv2.THRESH_BINARY)

    canvas = np.zeros((TARGET, TARGET), dtype=np.uint8)
    ox, oy = (TARGET - nw) // 2, (TARGET - nh) // 2
    canvas[oy:oy+nh, ox:ox+nw] = resized

    final_png = cv2.bitwise_not(canvas)
    final_display = cv2.cvtColor(final_png, cv2.COLOR_GRAY2RGB)
    hist_final = make_histogram_image(final_png, "Distribusi Pixel — Crop & Resize")
    final_active_mask = (canvas == 255)
    stats_final = compute_pixel_stats(final_png, active_mask=final_active_mask)
    df_final = arr_to_preview_df(final_png)
    info_resize = "Bounding box : {}×{} px\nPadding      : {} px\nSkala resize : {:.3f}×\nKanvas akhir : {}×{} px".format(
        w, h, pad, scale, TARGET, TARGET)

    # ── TAHAP 5.5: BACKGROUND REMOVAL ──────────────────────
    # Konversi RGBA yang aman secara struktur untuk UI Web
    rgba = cv2.cvtColor(final_png, cv2.COLOR_GRAY2RGBA)
    rgba[final_png == 255] = [255, 255, 255, 0]
    rgba[final_png == 0] = [0, 0, 0, 255]

    hist_transparent = make_histogram_image(rgba, "Distribusi Pixel — RGBA")
    transparent_active_mask = (canvas == 255)
    stats_transparent = compute_pixel_stats(rgba, active_mask=transparent_active_mask)
    df_transparent = arr_to_preview_df(rgba)

    checker = np.zeros((TARGET, TARGET, 3), dtype=np.uint8)
    cell = 20
    for cy in range(0, TARGET, cell):
        for cx in range(0, TARGET, cell):
            color = 240 if ((cx // cell) + (cy // cell)) % 2 == 0 else 215
            checker[cy:cy+cell, cx:cx+cell] = (color, color, color)

    alpha = rgba[:, :, 3:4].astype(np.float32) / 255.0
    rgb_fg = rgba[:, :, :3].astype(np.float32)
    transparent_preview = (rgb_fg * alpha + checker.astype(np.float32) * (1 - alpha)).astype(np.uint8)

    transparent_pixel_count = int(np.sum(final_png == 0))
    info_transparent = "Area objek    : {:,} px\nArea transparan: {:,} px\nDimensi       : {}×{} px (RGBA)".format(
        transparent_pixel_count, TARGET*TARGET - transparent_pixel_count, TARGET, TARGET
    )

    # ── TAHAP 6: EXPORT SVG ─────────────────────────────────
    path_svg = os.path.join(tmp_dir, "signature_vectorized.svg")
    dwg = svgwrite.Drawing(path_svg, size=("400px", "400px"), profile="full")

    svg_contours, svg_hier = cv2.findContours(canvas, cv2.RETR_TREE, cv2.CHAIN_APPROX_TC89_KCOS)
    path_count = 0
    if svg_hier is not None:
        for cnt in svg_contours:
            if cv2.contourArea(cnt) > min_area:
                approx = cv2.approxPolyDP(cnt, epsilon=epsilon, closed=True)
                pts = [(int(p[0][0]), int(p[0][1])) for p in approx]
                if len(pts) >= 3:
                    dwg.add(dwg.polygon(points=pts, fill="black", stroke="black", stroke_width=0.3))
                    path_count += 1
    dwg.save()
    info_svg = "Format      : SVG\nKualitas    : {} (epsilon={})\nPath/polygon: {}\nViewbox     : 400×400".format(
        svg_quality, epsilon, path_count)

    # ── SIMPAN FILE ──────────────────────────────────────────
    path_clean = os.path.join(tmp_dir, "signature_clean.png")
    cv2.imwrite(path_clean, final_png)

    path_transparent = os.path.join(tmp_dir, "signature_transparent.png")
    # OpenCV imwrite membutuhkan format BGRA secara native untuk disave ke disk
    cv2.imwrite(path_transparent, cv2.cvtColor(rgba, cv2.COLOR_RGBA2BGRA))

    path_csv = os.path.join(tmp_dir, "signature_matrix.csv")
    pd.DataFrame(canvas).to_csv(path_csv, index=False)

    path_zip = os.path.join(tmp_dir, "hasil_signature.zip")
    with zipfile.ZipFile(path_zip, "w") as zf:
        zf.write(path_clean, arcname="signature_clean.png")
        zf.write(path_transparent, arcname="signature_transparent.png")
        zf.write(path_svg, arcname="signature_vectorized.svg")
        zf.write(path_csv, arcname="signature_matrix.csv")

    # ── OUTPUT FINAL (ringkasan akhir) ─────────────────────
    final_summary_stats = (
        "Jumlah kontur valid : {}\n"
        "Ukuran hasil        : {}×{} px\n"
        "Jumlah path SVG     : {}\n"
        "Kualitas SVG        : {}"
    ).format(len(valid_contours), TARGET, TARGET, path_count, svg_quality)

    return (
        diagnosis_html,
        gray_display,    hist_gray,        stats_gray,        info_gray,    df_gray,
        thresh_display,  hist_thresh,      stats_thresh,      info_thresh,  df_thresh,
        morph_display,   hist_morph,       stats_morph,       info_morph,   df_morph,
        contour_display, hist_contour,     stats_contour,     info_contour, df_contour,
        final_display,   hist_final,       stats_final,       info_resize,  df_final,
        transparent_preview, hist_transparent, stats_transparent, info_transparent, df_transparent,
        final_display, transparent_preview, path_svg, final_summary_stats,
        path_svg, info_svg,
        path_clean, path_transparent, path_csv, path_zip,
    )

```
